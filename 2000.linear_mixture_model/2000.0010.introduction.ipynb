## 解説

### クラスタリングの評価

説明変数空間でのクラスタリングを行えることをすでに示しました。
場合によってはクラスタリングの評価指標からクラスタ数が決まります。
しかし、クラスタリングは教師なし学習なので、
クラスタリング結果それ自体では得られたクラスタが妥当かを評価する評価指標はありません。
評価する一つの方法はより定量的な評価指標を持つ教師あり学習と結びつけて、
教師データから得られる評価指標を用いて
クラスタリング結果の有用性を示すことです。



### 回帰関数の切り替え


![図：領域分割と回帰関数切り替え](image_keep/PQ-GMM.PNG)

図：領域分割と回帰関数切り替え

P,Q領域が完全に分離していれば図(a)のように領域ごとに回帰関数を作ることができます。

$$y^P = f^P(\vec{x})$$
$$y^Q = f^Q(\vec{x})$$

これはP,Q領域が完全に分離していれば,
領域P,Qを選択する関数$\pi^p(\vec{x})$=0 or 1を用いて
$$ y = \pi^P(\vec{x}) f^P(\vec{x}) + \pi^Q(\vec{x}) f^Q(\vec{x})$$
とすることと同じです。

一方、図(b)のようにP,Q領域が分離していない場合は、どうするでしょうか。
紫色の点はP,Q領域どちらにも属しているように見えます。
もし、領域P,Qに存在する確率関数$\pi^p(\vec{x})$=[0,1]、
（総和$\pi^P(\vec{x}) + \pi^Q(\vec{x})=1$）が評価できれば、
$$ y = \pi^P(\vec{x}) f^P(\vec{x}) + \pi^Q(\vec{x}) f^Q(\vec{x})$$
と回帰関数をなめらかに切り替える形に定式化することも可能です。
このために確率を評価できるクラスタリング手法を用います。


### 補足

結晶の全エネルギーを考える場合には原子環境説明変数をまず作成し、
原子環境説明変数に操作を加えて結晶説明変数を考えることが可能です。
結晶説明変数のクラスタリングを考え全エネルギー回帰に利用できます。

手法１．原子説明変数→結晶説明変数、
'結晶説明変数＋クラスタリング'→目的変数


もし、原子環境説明変数のクラスタリングの方が
結晶説明変数のクラスタリングによるクラス分離が良いとしたら
結晶説明変数よりも原子環境説明変数のクラスを使ったほうが回帰性能もまた良くなるのではないでしょうか。
この問題を行います。

手法２．'原子説明変数＋クラスタリング'→結晶説明変数→目的変数
