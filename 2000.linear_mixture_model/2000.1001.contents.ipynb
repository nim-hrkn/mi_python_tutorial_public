## 例題

Q. 解明しようしている材料問題（仮想問題）が何か？

A. 炭素８元素からなる結晶構造の高精度エネルギー予測を行う。

Q. データがどうようなものか？

A. 説明変数: GRRM法により構造探索をした炭素８元素からなる結晶構造をBehlerのsymmetry functionに変換した。

A. 目的変数: 結晶の全エネルギー

Q. どのような手法で何を抽出しようとするか？

A. GMM＋線形回帰により回帰性能をあげる。
