### 簡単な回帰データの作成

これはすでに基礎編：回帰で作成したデータとほぼ同じデータを作成します。


```python
import sys
import numpy as np
import matplotlib.pyplot as plt
```

```python
def makedata1(filename, xrange=(0.0, 8.0, 0.1)):
    """make data

    Args:
        filename (str): filename
        xrange (tuple, optional): range of X. Defaults to (0.0, 8.0, 0.1).
    """
    x = np.arange(xrange[0], xrange[1], xrange[2])
    x2 = x*x
    x3 = x**3
    x4 = x**4
    x5 = x**5
    y = -np.sin(x)  # +np.random.normal(0,0.05,size=x.shape)

    f = open(filename, "w")
    xarray = []
    f.write(",".join(["x1", "x2", "x3", "x4", "x5", "y"])+"\n")
    for t in zip(x, x2, x3, x4, x5, y):
        f.write(",".join(map(str, t))+"\n")
    f.close()
    print(filename, " is made.")
```

```python
g_filename = "../data_calculated/sin5.csv"
makedata1(g_filename)
```
