

## 例題


Q. 解明しようしている材料問題（仮想問題）が何か？

A. GRRM法により構造探索をした炭素８元素からなる結晶構造のうち最もエネルギーが低い構造をベイズ最適化により予測する。

Q. データがどうようなものか？

A.1.簡単な例

説明変数: $x^n$、sin関数

目的変数: sin関数

A.2.炭素構造

説明変数: 炭素８元素からなる結晶構造をBehlerのsymmetry functionに変換したもの

目的変数: 結晶の全エネルギー

Q. どのような手法で何を抽出しようとするか？

A.　ベイズ最適化を行う。
