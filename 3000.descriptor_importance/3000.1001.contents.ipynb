##  例題


Q. 解明しようしている材料問題（仮想問題）が何か？

A. 遷移金属(コバルト)ー希土類金属合金のキュリー温度を予測するモデルの説明変数の重要性を評価する。

Q. データがどうようなものか？

A. 説明変数： 構成元素と構造から作成した説明変数

A. 目的変数：キュリー温度

Q. どのような手法で何を抽出しようとするか？

A. ランダムフォレストにおける説明変数重要度、回帰手法によらない説明変数重要度の評価手法、線型回帰の回帰係数との比較を行う。
