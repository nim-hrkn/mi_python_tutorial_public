## 解説 

Association rule learning (関連性探索)、itemset mining、相関ルール探索などとも呼びます。

トランザクションが複数の検索条件（アイテム）から成っているとして、
このうちから頻出集合を発見するアルゴリズムです。

**利点**
とても高速に頻度、ruleを見つけることが可能です。
トランザクションが全てのアイテムの条件を持つ必要がありません。
データが大量にある場合に特に有効な手法です。
欠落したデータであっても、そのデータなりの頻出集合を発見します。

ref. 

https://en.wikipedia.org/wiki/Association_rule_learning

https://www.kamishima.net/archive/freqpat.pdf


例えば、以下の表を考えます。これは1900年頃までの、超伝導体の一部の物質の構成原子とそのTCを示します。
ref. https://en.wikipedia.org/wiki/Superconductivity

| トランザクション | アイテム |
|------|------|
| 1 | TC<30 Hg |
| 2 | TC<30 Pb |
| 3 | TC<30 NbC |
| 4 | TC<30 NbN |
| 5 | TC<30 Nb Sn |
| 6 | TC<30 V Si |
| 7 | TC<30 Nb Ge |
| 8 | TC>30 La Ba Cu O  |
| 9 | TC>30 La Sr Cu O | 
| 10 | TC>30 Y Ba Cu O | 
| 11 | TC>30 Bi Sr Ca Cu O | 
| 12 | TC>30 Tl Ba Ca Cu O | 
| 13 | TC>30 K C | 

TC>30は30K以上のTCを持つ。TC<30は30K以下のTCを持つことを示します。
例えば、8番目はLa Ba Cu Oから成る物質が30K以上のTCを持つことを示します


相関ルールでは
$$ X \Rightarrow Y $$
となるアイテムXの条件を探します。この条件Xは

* item1 AND item2 AND item3 AND ...

などANDによる単一条件式（ORが無い）で、Yもアイテムからなる単一条件式です。

上の例では、

{Cu O} $\Rightarrow$ TC>30

という条件です。Cu Oからなる物質は高い頻度でTC>30となることを示します。


また、分類の時に連続変数を検索条件に直しました。同様に検索条件をitemとすることも可能です。

例: 抵抗のしきい値を10 $ \Omega $として
- resistivity = 100 $ \Omega \rightarrow$ "resitivity>10$ \Omega $"というアイテムを持つ。
- resistivity = 0.1 $ \Omega \rightarrow$ "resitivity>10$ \Omega $"というアイテムを持つ。



### 評価指標

* support(X) = 条件Xを満たすトランザクション数/ 全トランザクション数

X= {Cu O}の場合、support(X) = 5/13 = 0.38

* confidence(X,Y) = 条件XとYを共に満たすトランザクション数/条件Xを満たすトランザクション数

Xを前提部(antecedent)、Yを結論部(consequent)と呼びます。

X= {Cu O}、Y=TC>30の場合、confidence(X,Y) = 5/6 = 0.83



### itemset mining利用の利点

1. トランザクション毎にアイテムの数が異なっていても条件を見つけられます。
2. 全ての組み合わせを計算するアルゴリズムがとても高速です。

回帰、分類、クラスタリング、次元圧縮を行うには**等長**の説明変数に変換せなばなりません。
古典ポテンシャルの場合のように非等長の生の説明変数を等長の説明変数に変換できる可能性はありますが、
一方、現実には生の説明変数が等長の場合の方が少ないでしょう。
itemset miningでは非等長の生の説明変数のまま解析を行うことができます。

### 補足

ここで使用するmlxtendのライブラリは基礎的な機能しかありません。

たとえば、下で示すように頻出マイニング結果が出力されます。

|id | support | itemset |
|--|-----|-----|
|1 | 1.000000 | (M_id==7) |
|2 | 1.000000 | (TC_id==5) |
|3 | 1.000000 | (M_id==7, TC_id==5) |
|4 | 0.947977 |(Mn, TC_id==5) |
|5 | 0.947977 |(Mn) |
|6 | 0.947977 |(Mn, TC_id==5, M_id==7) |
|7 | 0.947977 |(Mn, M_id==7) |

実際はsupport＝１が３：(M_id==7, TC_id==5)でその部分集合が１：(M_id==7)と２：(TC_id==5) 、
support=0.947977が６：(Mn, TC_id==5, M_id==7) でその部分集合が４：(Mn, TC_id==5)、５：(Mn)、７：(Mn, M_id==7)です。

より高機能なソフトに例えばLCMがあります。
LCMでは上の場合に、３：(M_id==7, TC_id==5)、６：(Mn, TC_id==5, M_id==7) のみを出力することが可能です。

LCM: http://research.nii.ac.jp/~uno/code/lcm.html
