LLMはプロンプトにより様々なシステムの構築が可能です。

# 物質科学への適用

物質科学のLLMの適用例を考えます。
機械学習を行う際に説明変数が必要です。
関係しそうもない説明変数を多数追加するより、関連しそうな他の物性値を説明変数に用いた方が少ない説明変数数かつより妥当な予測モデルが作成できるはずです。

このような物性値関連性探索システムに
「マテリアルキュレーション(C)（MatQ(TM))用物性間関連性探索システムにおける探索優先順位つけ」吉武道子、
ref. https://www.jstage.jst.go.jp/article/jsapmeeting/2021.1/0/2021.1_326/_article/-char/ja/
があります。
これは書籍上の物性値を人がアノテーションし、自然言語解析モデルを用いて関連性評価値を得ています。
このためのデータ作成は人がアノテーションするので膨大な手間と時間がかかります。
この手法は、書籍の一部をアノテーションして学習モデルを作成して、書籍の残りに対して適用し、物性値の間の関連性を得て、
書籍に記載された全ての物性値に対して関連性評価値を得ています。
そのため、妥当でない関連性評価値を得ている場合もあります。
また、説明変数間の関係を書いた文を示すことで具体的な関連性の根拠を示していますが、書籍の一部を表示することは著作権の問題があるでしょう。


## プロンプト作成方法

同様のシステムをLLMを用いて簡単に作成してみます。

Chatbotを作る場合に野口竜司氏考案の「文系AI式7Rプロンプト」が有用です。
(文献：「ChatGPT時代の文系AI人材になる」東洋経済新報社，野口竜司，ISBN-10 ‏ : ‎ 4492047395) 
このプロンプトは７つの要素からなります。


1. Request (依頼）を出す。
2. Role (役割）を決める。
3. Regulatin（形式）を指定する。
4. Rule（ルール）を定める。
5. Review & Refine（評価＆改善）を求める。
6. Reference（参照知識・例）を与える。

＋

7. Run Senario(実行シナリオ）で制御する。

これを利用した例が、例えば、以下で紹介されています。

ref. 銀行業務就任プレスリリースの文系AI式7Rプロンプト具体例　(2023/08/19)
https://www.youtube.com/watch?v=GCwEIvzxPSc&list=PLYM5HQYI7Lh11FICAoKFRD0eVc5y73iRT&index=38


```
# 依頼
あなたは{# 役割}です。次の{# ルール}を必ず守り、以下の「プレスリリース」への想定質問と回答例を上げて、{# 形式}の形式で出力してください。

# 役割
広報担当者

# 形式
- 表形式
- 列は「想定質問」、「回答例」「聞かれる可能性」

# ルール
- 想定質問と回答例を３つづつ挙げてください。
- {# 評価}の方法で評価し、出力に含めてください。
- {# 参照]を例にして、出力してください。
- {# 実行シナリオ}に沿って私とやり取りしてください。

# 評価
-　想定質問が聞かれる可能性を五段階で評価

# 参照
- なし

# 実行シナリオ
1. 一回目の出力を行う「どの案がよろしいですか？」と私に聞く。
2. 私が選んだ案の傾向を把握して、{# 依頼}を実行する。
3. 私に「このタスクを終了させてよろしいですか？」と聞き、終了してよければタスクを終了させる。終了していけなかれば再度{# 依頼}を実施する。
```

$$

ここで、$ L $ はウィーデマン＝フランツの法則の比例定数（ローレンツ数）で、温度が異なる場合でもほぼ一定の値をとります。

この法則は主に金属に適用され、電子が熱と電気の両方の主要な運搬者である場合に有効です。そのため、一般に金属で見られるように、電気伝導率が高い物質は熱伝導率も高い傾向にあります。

しかし、この関係は非金属や半導体には直接適用できないことが多いです。非金属材料では、熱は格子振動（フォノン）によって伝わることが多く、これは電気伝導とは異なるメカニズムです。したがって、非金属や半導体では電気伝導率と熱伝導率の間に単純な比例関係は見られません。
