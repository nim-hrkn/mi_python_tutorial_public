# 機械学習手法の紹介

##  機械学習手法の目的による分類

機械学習には回帰、分類、次元圧縮、クラスタリングがあります。

1. 機械学習手法の目的の一つは予測を行うことです。これは「予測問題」と
も呼ばれます。
予測問題は目的変数がある問題で，「教師あり学習」と
も呼ばれます．予測問題を達成する手法は回帰や分類です。

教師あり学習では，目的変数値と学習済みモデルによる予測値の一致度を用いて予測モデルの妥当性を明確に定義できます。定量的な一致度（評価指標）は後ほど紹介します。

2. 機械学習手法もう一つの目的は説明変数から法則性を見つけることです。
これは「記述問題」と呼ばれます。目的変数が無い問題設定であり，
「教師なし学習」と呼びます。
記述問題を達成する手法中には，次元圧縮やクラス
タリングがあります。

一方、教師なし学習では，妥当性を明確に定義することができません．例えば，クラスタリングでクラスター数に関して何かしらの評価基準を考えることはできるが，それはそもそも答え（目的変数値）が無い解析であり，ある評価基準値が与えられたとしても
それが何かの意味でクラスター数の妥当性を表せるのかは分かりません。



## データの規格化

教師あり、教師なし学習を説明する前に説明しておくことがある。
説明変数ごとにデータが持つ意味合いは異なっており，例えば物理量の測
定値に限った場合でも，単位が異なるものを同列に扱うことは適切ではない．
これを改善するための最も基本的な手順が各説明変数の回帰に及ぼす影響が
同程度になるように値を変換する「データ規格化」である．一般的な機械学習ラ
イブラリは規格化されたデータを想定して
開発されているので, これを行った方が良い．

データ規格化では

- Min-Max Normalization：[0,1] 区間への線形変換
- Z-score Normalization ：平均値 0，標準偏差 1 の分布とする線形変換

がよく用いられる ．規格化は，説明変数ごと変換関数を作成する場合が多
いが，説明変数ごとではなくデータ全体に対して変換する場合がある．
どの規格化が良いのかは，例えば，実際に回帰・分類モデルの性能を評価し
て判断する．

## 教師あり学習

機械学習の人は（二値）分類から入るのですが、物性の人は回帰から入った方が分かりやすいので先に回帰を説明します。

### 回帰

回帰にはよく知られた手法として

1. 線形回帰
2. ニューラルネットワーク回帰
3. カーネル回帰
4. ...

など様々な手法がありますが、
scikit-learnで利用可能な線形回帰とカーネル回帰をまず説明します。（ニューラルネットワーク回帰は説明しませんし例でも出てきません。）

#### 線型回帰

説明変数を表すベクトルを
$$ \vec{x}_i = \left( x_{i1} , x_{i2}, \cdots, x_{iP} \right) $$
と定義します。
これと同じサイズの係数ベクトル
$$ \vec{w} = \left( w_1 , w_2, \cdots, w_P \right) $$
と定義し、
回帰式
$$ f(\vec{x})= \sum_i^{P} w_p x_p + w_0 = (\vec{w}, \vec{x}) + w_0 $$
が与えられた時にコスト関数
$$L^{reg} = \sum_i^N (y_i - f(\vec{x}_i))^2 + \alpha |\!| \vec{w}|\!|_n^n $$
を最小化するように $\vec{w}$と$w_0$を決定することが回帰モデルの学習です。
回帰モデルが得られ、未知データ$\vec{x}^{new}$が与えられると、$f(\vec{x}^{new})$により、予測値を得ます。

（第一項には$1/(2N)$などの規格化が入ります。
https://scikit-learn.org/stable/modules/generated/sklearn.linear_model.Lasso.html?highlight=lasso#sklearn.linear_model.Lasso
)


この式の第二項は罰則項と呼ばれ，機械学習手法では頻繁に用いられます。
$|\!| \vec{w} |\!|_n$ はベクトル $\vec{w}$ の 「$n$ 次ノルム」と呼ばれる．更に，この m 乗を $|\!|\vec{w} |\!|_n^m$ も定義します。よく用いられるのは$|\cdot|$を絶対値として、

$$ |\!|\vec{w}|\!|_1^1= \sum_{p=1}^{P} |w_p| $$
$$ |\!|\vec{w}|\!|_2^2= \sum_{p=2}^{P} (w_p)^2 $$

です。
ここで$\alpha$は罰則項の大きさを決める値であり、ハイパーパラメタと呼ばれます。
その大きさは観測データを用いた評価関数の最適化の過程で決めることが良く行われます。


- 罰則項なし($\alpha=0$): (いわゆる）線型回帰

線型回帰は$L^{reg}$で与えられた罰則項の有無、種類で以下に分類されます。
- n=1 : Lasso
- n=2 : リッジ(Ridge)回帰

特筆すべきはLassoとリッジ回帰は大域最小解を与え、罰則項無しの線型回帰でしばしば問題になるる説明変数間の（多重）共線
性の影響を受けずに解を一意に与えることができます。
また，Lasso では罰則項は多数の説明変数の中から不要な説明変数の係数を 0 にする特徴を持ち、回帰と同時に説明変数選択を行うことができる手法です。


#### カーネルリッジ回帰

カーネル回帰関数は以下のような形を持ちます。馴染みが無いため意味が分からないかもしれませんが、係数$\alpha_i$に対して線形な関数系です。

$$
f(\vec x) = \sum \alpha_i K(\vec x\vec ,x_i)
$$

RBFカーネルと呼ばれる以下のガウシアンの形の関数が$K$としてよく用いられます。
$$
K(\vec x,\vec x') = \exp(-\gamma|\!|\vec x-\vec x'|\!|_2^2)
$$

この予測の模式図を以下に示します。
$|\vec x-\vec x'|\!|_2^2$により、
説明変数空間で距離が近いデータインスタンスの寄与が多くなりますが、
同時に$\alpha_i$（ここで$i$はデータインスタンスのindexである。）があります。この項は例えば、
存在すると回帰に合わない係数を小さくして回帰に影響を与えないようにする寄与があります。

![カーネルリッジ法](image_keep/kernelridge.png)

また、
$$
K(\vec x,\vec x') = \exp(-\gamma|\!|\vec x-\vec x'|\!|_1)
$$
とするラプラシアンカーネルもよく用いられます。

カーネルリッジ回帰では
以下の評価関数を最小化します。
$$
L = \sum_{i=1}^{N} |\!|y_i - f(\vec x_i)|\!|_2^2  + \alpha |\!|\vec w|\!|_2^2
$$
これはRidge回帰と同じく回帰係数L2ノルムの罰則項を持ちます。



RBFカーネルやラプラシアンカーネルではある$\vec x$点の回帰値として
$|\!|\vec x-\vec x_i|\!|_2$の値が小さいデータインスタンス$\vec x_i$が$K(\vec x,\vec x_i)$の寄与率で回帰モデルに用いられます。
また、距離依存性を無視し、$\vec x$のK個の説明変数近傍点の平均値を用いたk近傍法と呼ばれる回帰モデルもあります。


### 分類

（日本語だとclassificationもclusteringも分類と訳されるので、それらを分類とクラスタリングと呼ぶことにします。）


ロジスティック回帰は分類の代表的手法である．これは線形回帰の式 f(x)
を logit 関数
$$P(\vec x) =
\frac{1}
{1 + \exp( − f(\vec x))}$$
に作用させて 0 から 1 までの量に直す。
$P(\vec x)$ に作用させることで
$-\infty$ から $\infty$ までの値を取れ
る関数$f(\vec x)$ を 0 から 1 までの値を取るよう
に変換し，確率と解釈する．scikit-learnの罰則項があるロジスティック回帰では次式の評
価関数を最小化することで分類モデルを学習する．

$$ L^{cls} =  \left[ |\!| \vec w |\!|_n^n + C\sum_i \log( \exp( - y_i f(\vec x_i) ) + 1 ) \right] $$

分類クラスが二つである場合はあるクラスと予測される確率を $P(\vec x)$，別な
クラスとして予測される確率を $1 − P(\vec x)$ と解釈できる。
分類するクラスが複
数の場合は，例えば，理解しやすい手法としてはone-vs-rest 法ではそのクラスか否かという分類モデルをクラス回数個学習し，最も確率が高いクラスを予測値とする。
（OvRを用いるscikit-learnのLogisticRegressionCVは複数の最適化されたCを持ち、
単一のCを持つLogisticRegressionより分類性能がかなり向上している。）

OvR以外にも多値ロジスティク回帰手法が存在する。多値ロジスティク回帰の詳細についてはビショップの本などを参照してほしい。

### 類似度と距離

1 次ノルム，2 次ノルムと関連して「距離」，そして「類似性」は機械学習に
おいて重要な概念である．
類似性・非類似性はデータが似ている・異なって
いることを表し、それらを定量化した数学的な実装が距離である．様々な
距離が定義されているが、ここでは以下の代表的な距離について説明を行う。

- ユークリッド距離 (Euclidean distance)：L2 ノルムと同じ二地点間の距離
の定義である．最も馴染みがある距離の定義と思う．非負値を取り，値が
小さいほど類似性が高い．
- マンハッタン距離 (Manhattan distance)：L1 ノルムと同じ二地点間の距離
と同じ定義である．例えば，碁盤目に道がある都市の中を移動する際には
直線距離（ユークリッド距離）では移動できない．道に沿って移動するた
めに，マンハッタン距離の方がタクシー運賃などを決める距離の定義とし
て適切である．非負値を取り，値が小さいほど類似性が高い．
- コサイン距離 (Cosine distance)：二地点をベクトルとして，ベクトルのコ
サインを計算する．–1 から 1 まで値を取り、値が大きいほど、類維度が高い。


これらの距離実装を知ると
カーネルリッジ法で説明したRDF kernelではデータインスタンス間のユークリッド距離の二乗が用いられていることが分かり、値が大きいほど類似度
が小さくなる距離として
$$
K(\vec x,\vec x') = \exp(-\gamma|\!|\vec x-\vec x'|\!|_2^1)
$$
また、異なる距離実装として
$$
K(\vec x,\vec x') = \exp(-\gamma|\!|\vec x-\vec x'|\!|_1^1)
$$
をなど用いることができることを理解できると思います。
また、目的変数に周期性があると分かっている観測データには、その周期性も取り入れてカーネルを作ることができます。
例えば、年月毎のデータで、異なる年でも月の依存関係が大きい場合は、同じ月では距離が近くなるようなカーネル関数を選ぶということです。

### if文による回帰・分類モデル

上で説明した回帰モデルは目的変数の説明変数空間上での超局面を求めているとも考えられます。

異なる考え方として
```
if x_i >= value1 then
  if x_j > value2 then
      y = yvalue1
  else
      y = yvalue2
  endif
else # x_i<value1
  if x_j > value3 then
      y = yvalue3
  else
      y = yvalue4
  endif  
endif
```
というif文を用いて予測値を得る回帰・分類モデルもよく用いられ
決定木と呼ばれます。決定木の上位の変数ほど重要性が高くなる重要度を定義することができます。

この決定木手法は目的変数予測値がなめらかに変化せずに、step関数的に変化する（微分値が０もしくは無限大）ことがif文による予測値の構成法により理解できると思います。

## 教師なし学習

### 次元圧縮

次元圧縮(次元集約、次元削減とも呼ばれます。)では
説明変数の行列$X$をそのまま用いるPCAなどと距離行列に直してから解析する多様体学習がよく用いられます。



### 特異値分解

まず、特異値分解を説明します。
特異値分解は
一般に$N\ne P$の非正方な行列$X$（$N,P$)を

$$ X = U S W^T$$

と分解します。
(pythonではscpyの関数scipy.linalg.svdで行うことができます。)
行列$S$は（$N,P$）の対角行列となります。
たとえば、$N>P$の場合には縦長の行列

$$
S = \left( 
 \begin{array}{c}
  \begin{array}{cccc}
   s_1 & & & \\
    & s_2 & & \\
    & & s_3 & \\
   & & & \ddots \\
   \end{array} \\
    0
  \end{array}
   \right)
$$

になります。
$s_i$は非負値に取ることができ、大きい順に並び替え($W$,$U$の要素も並び替える）ておきます。


大きな幾つかの$s_j$もしくはあるしきい値以上の$s_j$だけを残し、それ以外の対角要素を０、例えば

$$
S' = \left( 
 \begin{array}{c}
  \begin{array}{cccc}
   s_1 & & & \\
    & s_2 & & \\
    & & 0 & \\
   & & & 0 \\
   \end{array} \\
    0
  \end{array}
   \right)
$$

とすることで次元圧縮し再びXの行列次元に再構成した行列（低ランク近似した行列、低ランク行列と呼ぶ）を$US'W^T$として求め解析することがあります。

また、
$$ X = U S' W^T$$
は(i,k)行列要素を書くと

$$
S_{ik} = \sum_j s_j (U_{ij}W_{jk})
$$

となります。これは行列$S$を$U_{ij}W_{jk}$からなる行列で展開していて展開系数が$s_j$であるとみなせます。つまり、上の場合は
$$
S_{ik} =  s_1 (U_{i1}W_{1k}) + s_2 (U_{i2}W_{2k}) 
$$
という展開とも理解できます。


### 主成分分解（PCA)

次元圧縮では主成分分解が最も有名でしょう。
分散を考えるために**平均値を引いたX**に対して$P\times P$の不偏共分散行列

$$\Sigma = \frac{1}{N-1} X^T X$$

を計算し、固有値分解します。ここに$N$はサンプルサイズ（データインスタンス数）です。
（なお、実際は規格化をし直すので規格化因子$1/(N-1)$は問題ではありません。）

これは
$$\Sigma = V \Lambda V^T $$
と固有値分解できます。
ここに、
$\Lambda$は非負値を持つ対角行列です。 
特異値分解の時と同様に$\Lambda$の対角項を大きい順に並び替え（Vも並び替える）、
説明変数のサイズより小さいサイズのVの部分行列で説明変数空間の部分空間を定義します。
$\Lambda$の対角項の和を１に規格化した対角成分を寄与率と言います。その和を累積寄与率と言います。

scikit-learn, solver='full'では
上で行った特異値分解を行い、

$$\Lambda = \frac{1}{n-1} S^T S$$

とします。次元圧縮後のベクトル$U'$は $S$ の一部の対角要素を０にした行列$S'$を用いて

$$ U' = X' W = (U S' W^T) W = U S' $$ 

として求めます。

### 多様体学習
多様体学習 (Manifold Learning) でよく用いられる手順は説明変数行列か
ら一度データ間の距離行列を計算し，それを活用して，データ空間を低次元
空間に圧縮変換します．この際に距離を活用する範囲を制限することで有用な
特徴を引き出す手法もあります。

本書では次元圧縮手法として PCA と多様体学習の例として t-SNE を用い
ます．PCA は大局的な構造が保たれるが，t-SNE は距離を活用する範囲を制限するため局所構造しか保たれない変換を行うという違いがあります．

### クラスタリング

クラスタリングの目的は類似なデータを同じクラスターに，非類似なデー
タを異なるクラスターに区別することです。この時に類似性，非類似性の
定量値として距離を用います。クラスタリング手法には，クラスター間の関係
に関して大きく分けて

1. 各クラスターが独立である手法
2. クラスター間に系統的な関係がある手法

があります。

####  各クラスターが独立である手法
1. 独立な各クラスターに区別するクラスタリング手法の代表である k-
Means 法は，決められたクラスター数 (K) の基で、
クラスター $C_k$ に属する 
$x^{( k )}_i$
と中心値 $\mu_k$を定めます。
そして、何かの定義の距離 $|\!| \cdots |\!|$ を用いた評価関数 
$$ J = \sum_{k=1}^K \sum_{n=1}^N |\!| x_i^{(j)} - \mu_k |\!|^2$$
を最小化するように
$x^{( j )}_i$
と $\mu_j$ を繰り返し更新し，最終的にクラスターを決定します。
この式を見ると分かる通り等法的な分割を行います。

2. 一方、非当方的な分割を行う手法として混合ガウスモデル(gaussian mixture model)があります。これは非当方的な分布を表すことができる
多変量正規分布
$$ N(\vec x| \vec \mu , \mathbf{\Sigma})  = \frac{1}{(2\pi)^{P/2}|\mathbf{\Sigma}|^{1/2}} \exp\left(-\frac{1}{2}(\vec x - \vec \mu )^T \mathbf{\Sigma}^{-1}
(\vec x - \vec \mu ) \right)
$$
を用います。
全体のデータ分布を
クラスタ$k$の分布の
平均値$\vec \mu_k$と分散共分散行列$\mathbf{\Sigma_k}$を用いた
それぞれの分布の確率の和
$$p(\vec x) = \sum_k^{K} w_k  N(\vec x| \vec \mu_k , \mathbf{\Sigma_k}) 
$$
$$ \sum_k w_k = 1$$
と表す式を用います。
それぞれの$\vec x$で規格化し直すと、
$\vec x$でのクラスター$k$の寄与率も求められます。

![図：GMM](image_keep/methods/GMMfig.PNG)
図：ガウス混合分布。一次元として書いた場合の頻度(box)と確率(線）を示す。

#### クラスター間に系統的な関係がある手法

クラスター間に系統的な関係がある手法の代表は
階層クラスタリング法です。
これはある距離実装を用いて
距離が近い順にクラスターを階層的に逐次生成します。クラスタ間の表示も階層的にされます。
二つ以上の要素が含まれるクラスター間の距離には様々な定義があります。


なお、教師なし学習のこの節ではデータインスタンス方向に行うようにクラスタリングの説明を行ったが、説明変数方向にクラスタリングを行うことも可能です。
