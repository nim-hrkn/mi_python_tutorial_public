# LLMを用いたコード生成


## 自然言語によるコーディング

GPT4oはプログラムを生成できます．その際に，指示において不明な部分を「よくあるコード」で補完します．この補完はユーザーの意図と異なることがよくあります．
プロンプトに詳細が少ないとGPT4oが補完する部分が多くなり，ユーザーの意図と異なるコードができることが増えます．
GPT4oにユーザーの意図通りのコードを生成させるためには過程も含めた詳細な指示を追加する必要があります．
当然その過程を人が明確に理解してプロンプトに記載する必要があります．
過程が複雑になるほど明確かつ詳細に指示を書くことにより，LLMを人が導く必要があります．
詳細なプロンプトは最初に人の作業量は多いですが，生成後は人の修正は少なくなり，結果的に作成者の意図通りのコードを短時間で生成することになるでしょう，逆に，簡単で大雑把なプロンプトは生成後に人が妥当かどうかを判断しながら，多くの部分を修正する必要があるでしょう．
これを図\ref{fig:AI-coding}にまとめました．
従来はユーザーが計算機言語によりプログラムを作成していました．図では計算機言語コードディングと書いています．
一方，LLMの進化により，計算機言語に依存せずに自然言語でアルゴリズムを書くことも可能です．図では自然言語コーディングと書いています．
人の計算機言語コーディング比率は前者が大きく，後者は少なくなります．一方，人の説明比率，LLMの計算機言語コーディング比率は前者から後者へ大きくなります．
現在は人の計算機コードディング比率が高い「AI支援コーディング」である場合が多いのですが，LLMが進歩すると自然言語コーディングの比率が増えた「AI駆動コーディング」，更に将来は全てを自然言語で記載する「全AIコーディング」が可能になると思われます．\footnote{AI支援コーディング，AI駆動コーディング，全AIコーディングは一般的な名称ではなく，著者が説明のために用いた用語です．}しかし，全AIコーディングが可能になるとしても，何をしたいか理解しているのはユーザーです．
将来のLLMは高度な推論やLLMが生成した内容の整合性が取れるようになることで，ユーザーの説明入力を少なくすることはできると思われますが，
ユーザーがLLMに不足なく説明できる必要があることは変わりません．

## 指示書の作成

プロンプトへのコード生成では何度も修正依頼を行えるため、実感しにくいことが少ないのですが，
GPT4oにコードの作成を依頼することはコードの外注をするようなものです．
外注先にコードの作成を依頼する場合に，大雑把な指示だと依頼された業者も何を行えば良いのか困りますし，納品されたコードは意味をなさないことが多いでしょう．
本書では図\ref{fig:AI-coding}の「AI支援コーディング」レベルでなく「全AIコーディング」を目指してプロンプトを書き，コード生成後にコードの不足分をユーザーが直にもしくはプロンプトを書き直しGPT4oに修正させる手法を取ります．
後者のGPT4oが修正する場合はユーザーは計算機言語を読めればよく直に書く必要性はかなり少なくなります．

本書ではコード例がある場合は，

1. GPT4oにサンプルコードを説明してもらい．
2. ユーザーがその説明を微修正しコードを生成するプロンプトを書き
3. GPT4oにコードを生成させます．

\noindent
\ref{sec:denoise}節のデノイズの問題ではサンプルコード無しに，

1. GPT4oにデノイズ手法の概要を説明してもらい．
2. 具体的なアルゴリズムを選択し，GPT4oに指示書プロンプトを作成してもらい，
3. GPT4oにコードを生成させます．
