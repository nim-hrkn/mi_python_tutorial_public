# 予測

原理・法則が分かっている場合の演繹法による予測と帰納法による予測の説明を行います。

## 問題：結晶構造から全エネルギーを予測する

### 事前に支配法則が分かっている世界での予測

多くの結晶構造とその全エネルギー($E$)が与えられた場合に$E$を予測する問題を考えます。
結晶構造は、
- 周期構造をabc軸長さとそれらの角度$\alpha$,$\beta$,$\gamma$で定義し、
- 更に周期内の元素と原子位置をを定義する

ことにより一つの結晶構造を定義します。
第一原理計算プログラムはこれらの構造を用いて、密度汎関数法を用いたディラック方程式やシュレディンガー方程式という「法則」を表現した式を計算して$E$を求めます。
このように、もし、世界を支配する方程式を知っていれば，計算機の中でつくる仮想世
界が，現実と寸分も違わない$E$を
$$ E_i = f(結晶構造_i) $$
により“予測”することができます。また、法則から結論を導き出すことを演繹的アプローチと呼びます。

$E$は得られた結晶構造を用いて「計算する」ことで得ますが、
一般的には、実験値による測定や観測で値得る、また法則による計算機実験で値を得ることを、以下では総称として「観測する」と書くことにします。
$N$個の結晶構造と全エネルギーを観測すると、この数値データは以下のようにテーブルの形でまとめることができます。

| 結晶構造ID| 結晶構造 | 全エネルギー |  
|---|---|---|
|ID$_1$| 結晶構造$_1$ | $E_1$ |
|ID$_2$| 結晶構造$_2$ | $E_2$ |
|ID$_3$| 結晶構造$_3$ | $E_3$ |
|...| ... | ... |
|ID$_N$| 結晶構造$_N$ | $E_N$ |

表: 結晶構造と全エネルギー

これら観測データの一構造（一行）をデータインスタンスといいます。
構造が$N$個あると、観測データのデータインスタンス数は$N$です。

（コメント：標本という言葉を用いると、上の数は標本の大きさ（sample size)です。標本数(number of samples)とは違います。）

## 事前に支配法則が分かっていない世界での予測

一方、事前に支配する法則が分かっていない世界では“予測”ができるでしょうか。
この場合はデータが大量にあれば、何らかの意味で「内挿」を行い予測でできることが期待されます。つまり、$obs$を上付き添え字として加えた大量の観測値データ
$$ (結晶構造_i^{obs}, E_i^{obs})$$
があれば、関数$f$
$$ E^{obs} \sim f(結晶構造^{obs})$$
を得ることができ、まだ観測していない$結晶構造^{new}$に対しても
$$ E^{new} = f(結晶構造^{new})$$
により妥当な$E$の予測値、$E^{new}$が得られることが期待されます。

この関数$f$を得る過程を関数𝑓を「学習する」と呼びます。そして、この関数𝑓を「予測モデル」と呼びます。 またこの関数は法則そのものではないのですが、予測が可能という意味で、法則のようなものを得たことになります。 
ここで、一般的な問題を考えるために、前者を$\vec x$、後者を$y$と書くことにし、
$\vec x$部分を**説明変数**（もしくはデスクリプタ）、$y$部分を**目的変数**と呼びます。

さて、ここでこの手法を実現する為の問題点があります。
何らかの意味で「内挿」するとは具体的にはどう関数$f$を見つけるのでしょうか。
まず、観測データで全観測点を通るように内挿した場合に何が起きるでしょうか。ある説明変数の変化に対して、観測誤差があるのが常なので、全観測点を通るように内挿した場合にガタガタ、ギザギザしたカーブができてしまいます。これでは多くの場合に良い予測ができません。データから、例えば、ニュートンの運動方程式、ケプラーの法則のような簡単な関係式を求めるのは不可能でしょう。
このため、誤差の影響を少なくし多数データの持つ関係を得るために、目的変数への**相関**が高い**予測モデル**
$$ y \sim f(\vec{x})$$
を求めます。具体的な求め方は後述します。（$y$もベクトルになることもありますが、今は簡単のためスカラとします。）



$E$のような「連続変数」の目的変数に対して$f$を求めることを「回帰」と呼びます。
一方、$E$がしきい値$E_0$以上か否かという問題設定にするとyes|noの「カテゴリー変数」になります。目的変数がカテゴリー変数の場合に$f$を求めることを「分類(classification）」と呼び、それらの関数を回帰モデル、分類モデルと呼びます。

このように観測データから目的変数と説明変数から作る高い相関を示す関数を発見するアプローチを帰納的アプローチと呼び， 特に，データだけから法則を発見することをデータ駆動型アプローチとも呼びます。


結晶から全エネルギーを予測する問題に戻ります。
一般に結晶内にある原子は結晶毎に異なります。これを踏まえて先の表を具体的に元素、原子位置に具体的に書き直してみます。
結晶構造ID$_1$には原子が1つ、結晶構造ID$_2$には原子が2つ、結晶構造ID$_2$には原子が3つ、あるとしてます。

| 結晶構造ID| 結晶格子 | 元素・原子  | 全エネルギー |  
|---|---|---|---|
|ID$_1$| $a_1$,$b_1$, $c_1$, $\alpha_1$, $\beta_1$, $\gamma_1$ | ((元素$_{1p}$,原子位置$_{1p}$))| $E_1$ |
|ID$_2$| $a_2$,$b_2$, $c_2$, $\alpha_2$, $\beta_2$, $\gamma_2$| (元素$_{2p}$,原子位置$_{2p}$),(元素$_{2q}$,原子位置$_{2q}$)| $E_2$ |
|ID$_3$| $a_3$,$b_3$, $c_3$, $\alpha_3$, $\beta_3$, $\gamma_3$ | (元素$_{3p}$,原子位置$_{3p}$),(元素$_{3q}$,原子位置$_{3q}$), (元素$_{3r}$,原子位置$_{3r}$)| $E_3$ |
|...|...|...|...|

表: 結晶格子、元素原子位置と全エネルギー

ここで$y\sim f(\vec x)$を求める際に2つの困難があります。
一つは、物理化学的な結晶の定義の問題です。
例えば、結晶格子を定義する($a_1$,$b_1$, $c_1$, $\alpha_1$, $\beta_1$, $\gamma_1$) と原子位置の軸を入れかえ、同時に原子位置の表現も入れ替えると同じ結晶構造を表します。また、格子サイズを倍にすると同じ結晶を定義したことになります。更に同じ元素の原子位置の並び順を入れ替えても実際は同じ結晶構造をします。このため、上の定義は結晶構造と全エネルギーを相関として関係づけるのには問題があります。
もう一点は、機械学習側に$\vec x$のサイズが等長でなく、様々なサイズの$\vec x$に対応する関数$f$を求めるのが困難であるという問題があります。
このため、非等長の結晶格子と元素・原子位置という$\vec x$の定義ではなく、物理化学的な条件を満たし、同長さのベクトルとして結晶構造を定義することが望まれます。
では、どのような定義があるでしょうか。この問題の場合は、機械学習古典ポテンシャルとして研究がなされ既に様々な定義がされてます。
例えば、X線回折や中性子実験で観測される動径分布関数（RDF)は、結晶格子や元素・原子位置の並び順と無関係に結晶構造を定義することができ説明変数として用いることができます。（より高精度に求めるには三体相関も取り入れます。には観測誤差が含まれる動径分布関数から複雑な結晶構造を求めることはとてもむずかしい問題です。）

$\textrm{𝑅𝐷𝐹}(𝑅_1), \textrm{𝑅𝐷𝐹}(𝑅_2), \textrm{𝑅𝐷𝐹}(𝑅_3), \cdots$に対する変数$x_{ij}$を**加工済み説明変数**と呼ぶことにします。
これで、物理学的な条件および、関数$f$を求めるための等長ベクトル$\vec x$を定義することができました。

機械学習では、Nデータインスタンスある（加工済み）説明変数$\vec{x}_i^{obs} =(x_{i1},x_{i2}, \cdots, x_{iP})$と目的変数$y_i^{obs}$に対して、相関から
$$ y \sim f(\vec{x})$$
となる関数$f$を学習します。そして、
新規結晶構造から作成される加工済み説明変数$\vec x^{obs}$に対して、得られた予測モデル$f$を適用し、予測値$y^{new}$を求めます。
