# 説明変数の特徴の見い出し方

予測モデルを求める具体的な手法を説明する前に、
表３のような *大量のデータ* から何を見出すことができるかを考えます。

## 特徴を表す説明変数が分かっている場合の解析

大量のデータがある場合に何ができるでしょうか。
説明変数がRDFを重ねてプロットする図３のような解析（可視化）が可能です。


以下では$𝑅_1,𝑅_2,𝑅_3$のみ考えることにします。
このため、線は単なるガイドと見てください。
この描き方の欠点は、描く構造（線）が多いと、線が重なって構造毎の変化が見えないことです。

次に、（以下の図４は、上の図３とは完全には対応しませんが、）多くの人は平均値と標準偏差を計算するでしょう。
![図：RDF, 平均値と標準偏差](image_keep/RDFfig/RDF_mean_stddev.PNG)
図、平均値と標準偏差

図(a)は平均値を示します。この図からはRDFの平均値の範囲と値が大きな$R_i$、小さな$R_i$が存在することが分かります。
図(b)は平均値と標準偏差を示します。この図からは$\sigma$が大きい$R_1$と$R_3$, $\sigma$が小さい${R_2}$があるデータであることが分かります。

では、$\sigma$が大きい$R_1$と$R_3$の間で何か相関があるでしょうか。例を次に示します。
![図：RDF, AとBの場合の値のscatter plot表示](image_keep/RDFfig/RDF_caseAB.PNG)
図、RDF,  AとBの場合の値のscatter plot表示

![図：RDF, AとBの場合の値のデータテーブル表示](image_keep/RDFfig/dataframe_caseAB.PNG)
図、RDF,  AとBの場合の値のデータテーブル表示

- 図(A)は$x_1(=RDF(R_1))$が大きいと$x1$も大きい、つまり$x_1$と$x_3(=RDF(R_3))$に正の相関があります。
- 一方、図(B)は$x_1$が大きいと$x_3$は小さい、つまり$x_1$と$x_3$に負の相関があります。
- 更にどちらも混じっている場合(C)もあります。

これらの場合はどのように判別できるでしょうか。
これは、($x_1, x_3$)を書くと(A)と(B)を容易に区別できます。
![図：case A and B, $x_1$ vs $x_3$](image_keep/RDFfig/RDF_AvsB.PNG)
図：AとBの場合の $x_1-\bar{x}_1$ vs $x_3-\bar{x}_3$

図では($x_1, x_3$)の代わりに($x_1-\bar{x}_1, x_3-\bar{x}_3$)を書いており、
- (A)の場合は$(x_1-\bar{x}_1)\times (x_3-\bar{x}_3)>0$の領域にデータがあり、
- (B)の場合は$(x_1-\bar{x}_1)\times (x_3-\bar{x}_3)<0$の領域にデータがあります。
- (C)の場合は図示してもこららの分布の偏りが見えません。

（これらのに変数の間の相関を定量的に表す数学式もありますがここでは使いません。）


図では(A)と（B)の場合を考えましたが、実際のデータでは(A)と（B)のどちらも混じっているでしょう。
では、どのような回帰モデルがありうるでしょうか？ $𝑎_1>0, 𝑎_3>0$として、


$$ 𝑦=𝑎_1 𝑥_1+𝑎_3 𝑥_3 $$もしくは
$$ 𝑦=-𝑎_1 𝑥_1-𝑎_3 𝑥_3 $$
は$x_1$と$x_3$が正の相関を示す変化をする場合に$y$が大きく変化している場合にありそうです。

$$ 𝑦=𝑎_1 𝑥_1−𝑎_3 𝑥_3 $$もしくは
$$ 𝑦=-𝑎_1 𝑥_1+𝑎_3 𝑥_3 $$
は$x_1$と$x_3$が負の相関を示す変化をする場合に$y$が大きく変化している場合にありそうです。

$$ 𝑦=𝑎_2 𝑥_2 $$ もしくは
$$ 𝑦=-𝑎_2 𝑥_2 $$ 
は(A),(B)の場合に$y$がほとんど変化していない場合にありそうです。

これらのうち相関が高いモデルが回帰で求まります。


上では$𝑅_1,𝑅_2,𝑅_3$が分かっていると仮定しているので図を書け、更に(A),(B),(C)をだけであると過程したので$f$の具体的な形の予想ができるのですが、では説明変数Xの特徴よく表す少数の$𝑅_𝑖$はどう見つけるのでしょうか？

1. $(𝑅_𝑖,\textrm{𝑅𝐷𝐹}(R_𝑖))$は１次元図として書けました。 $(𝑅_𝑖,\textrm{𝑅𝐷𝐹}(R_𝑖))$の図があっても、曲線が複雑な($\sigma$が大きな$R_i$が多いなど）場合はどの$𝑅_𝑖$が最も特徴的なのかが分からない。
2. さらに、説明変数$x_i$が多次元の場合はそのそもRDF図のようななめらかに変化する図が書けません。

主として以上の困難があり、説明変数の特徴をよく示す説明変数（もしくは変換した説明変数）を人間が見つけるのは困難です。

このために、「次元圧縮」という手法を用いて説明変数$x$ (上の行列X)を低次元化し、変数の特徴を求めることがよく行われます。うまく次元圧縮を行うと、データインスタンス間の相関がよりうまくとれるようになり、結果として
回帰、分類性能が向上し、クラスタリングにおいてもより孤立したクラスーたを得ることができる場合があります。
また、次元圧縮は説明変数間に関係があると仮定し、それらの間の擬似的な法則を求める場合にも有用です。（推薦システムの例で出てきます。）


また、説明変数空間を分割する「クラスタリング」とは回帰・分類モデル学習時に有用です。

図(a)のように説明変数空間で連続してデータインスタンスが存在する場合は一つの回帰モデルを作ることができます。
一方、図８(b)のようにデータインスタンスが説明変数空間で十分に分離していることが「分かれば」、分離したデータ毎に回帰をすればより狭い説明変数空間での学習いなり、より簡単に回帰予測モデルを学習することもできることが期待されます。このデータインスタンスがP領域とQ領域へ分離しているかの解析は（次元圧縮を行わない高次元空間でも）クラスタリングで行えます。（P領域とQ領域がよく分離していない場合が回帰モデル切り替え例で出てきます。）
