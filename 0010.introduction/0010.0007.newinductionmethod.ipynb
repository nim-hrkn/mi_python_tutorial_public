# 新帰納法による知識獲得

## 科学の歴史

迂遠ですが、科学の歴史を簡単に述べます。

１．帰納法：
データから個人のひらめきにより基礎法則や原理を発見しました。
例）ケプラーの法則、メンデレーフ周期律表。

２．演繹法：
原理から方程式を解く理論中心の科学です。
例）学校で勉強するほぼ全ての科学です。例，解析力学，統計力学，電磁気学，量子力学を学び，演習で手で解いたはずです．

３. 計算科学：
計算機を用いたシミュレーションにより予測・再現する科学です。
例）Gaussainなどの第一原理計算、計算機によるシミュレーション。

4．データと計算機を利用した「新」帰納法：スマート化とも呼ばれる。
データ処理による“法則”（相関）の発見を行います。
例）本書で初歩を勉強します。

## 計算機を用いた新帰納法

帰納法では仮説（例えば、方程式）を立てて実験により原理・法則を検証します（下図左）。
新帰納法ではこれを計算機が大量に行います。（下図右）ただし、新帰納法では現状では「相関」を求めます。このため下図右では発見されるものは「法則のようなもの」と書いています。また法則は複数かもしれません。
演繹法主体の世界と異なり、既存知識からこの仮説は正しいはず、という考え方はしません。計算機を用いた試行錯誤で知識獲得を行います。

## 帰納法の考え方に慣れること

演繹法の世界（理論物理や理論化学）と帰納法の世界（情報理論）とは大きく異なります。
演繹法の世界では原理を記述した方程式から導き出される最適なやり方が存在します。
理論物理や理論化学などの演繹法の考え方に慣れていると、情報理論の扱いにおいても最適なやり方があると勘違いしてしまいます。
たとえば、説明変数をどう選ぶのかという質問がよくあります。
おそらく、演繹法の考え方に従い何かしら最適な説明変数選択法があるのだと勘違いしているのだと思います。
説明変数はいろいろと試してみてその問題に対して適切なもの（最適とは限らない）を選択するという手順を踏むしかありません。
情報理論においては原理は無いため最適なやり方は対象により異なります。
目的を達成するために幾つか（幾つもの）の手段で試してみることが必要です。

学生は教科書により学問を学びます。
教科書というのは理論（手段）をまとめたものなので、教科書を用いて勉強すると目的より手段が先に来てしまいます。
大学で理論が分かる賢い人ほど手段を学ぶ傾向が強くなってしまいます。
情報理論というのは物性物理学に比べると実学であり目的に応じて数学・統計の知識を用いて様々な手段を研究・実行する学問です。
目的が最初に存在して、目的を実現するための手段を選択します。
賢い情報理論学者は様々な手法を知っていて、目的に応じて最適な手法を選択します。

理論物理や理論化学など演繹法の世界では基礎方程式（学理）が存在し、厳密解、最適解を出すことを求められ、厳密解に近い解を与える近似解法が良い解法とされます。
一方、情報理論は演繹法の世界のこれを解けば絶対に正しいという基礎方程式が無い世界です。このため手段は全て近似手法です。
しかし、例えば、alphaGoや自動運転の技術は決して厳密解、最適解を出しているわけではありませんが、人間よい強い碁を打つ、事故を起こさず車を移動させるという目的を人間にとって役に立つレベルで実現しています。
帰納法を扱う考え方に慣れると理解に戸惑わなくても済むはずです。そのための手助けをこの本で行いたいと思います。
