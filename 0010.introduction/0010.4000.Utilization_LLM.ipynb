# LLMの利用

## LLMを用いた学習

多くの方は学校に通って学習していた（知識を得た）と思います．
伝統的な講義形式の教育は図\ref{fig:LLM_learning_change}(a)の形と思います．
この図は「教科書・講義」，「学習者（生徒）」，「講義での質問・演習」との関係を示しています．
学習者が教科書・授業から知識を得ます．
そして，学習者が知識を確認したい，また更に知識を得たい場合には質問を通じて行います．
残念なたら，この質問は多くの場合は授業時間や休み時間のごく短い時間に限られます．
授業を受けた多くの学生は理解できない事項があった場合にフラストレーションを持ったと思います．
また，学習者が演習問題を解くことで学習者の理解度を測り方法も古くから採用されています．これは現実の問題解決の練習をするためです．しかしながら，提供される問題の範囲が限られていることや，過去の問題に依存しているため，学習者が実際に直面する現実の問題と大きく異なることがよくあります．

一方，LLMは非常に柔軟かつ強力な支援を行えます．これを図\ref{fig:LLM_learning_change}(b)に示しました．
LLMは教科書の内容を大きく超える広範な質問に対しても回答する能力があります．
これにより学習者は指定された教材の範囲に限らずその領域を超えた内容まで知識を獲得することができます．
さらに，LLMは学習者が直面する実際の問題解決を問題毎に支援できます．現在最も上手くできた例は計算機言語のコード生成でしょう．

## プロンプトエンジニアリング

LLMでは，ユーザーが実行してほしいタスクや質問の具体的な指示を含むテキストを「プロンプト」と呼び，このプロンプトを用いてLLMとやり取りします．ChatGPT公式の「プロンプトエンジニアリング」ガイドには，より関連性の高い回答を得るための戦術が紹介されています．このガイドはOpenAIのWebページ（OpenAI Platform\footnote{2024年10月時点では https://platform.openai.com/docs/guides/prompt-engineering}）で確認できます．この解説とプロンプト例はLLMをより深く理解し，上手に活用するための貴重な資源です．ただし，GPTに適切な回答をさせるための手法は一つではなく，多くの非公式の「プロンプトエンジニアリング」ガイドが存在しています．
プロンプトエンジニアリングでは，以下の点を考慮します．

1. 文章を明確にする．
2. 曖昧さを排除し，具体的な情報を提示する．
3. 複雑なタスクはステップごとに分けて提示する．
4. 背景情報を提示する．
5. 回答を評価し，プロンプトを修正する．

1と2に対応する方法として，複数の例を示すfew shotプロンプトがあります．例示を行わないのをzero shotプロンプトと呼びます．私はzero shotプロンプトが意図と異なる場合に，few shotプロンプトを使用します．
しかし，これらを全てを満たすプロンプトを人が作成することは困難です．
\footnote{これに対応するため，ChatGPTでも最近プロンプトジェネレーターを実装しました．}
本書では，ユーザーの負担を減らすために以下の方針を取ります．

### 質問方向

LLMの主な使い方は，ユーザーが興味を持つ事項について詳細を尋ねること（知識深化）と思います．
しかし，ある事項の上位概念や類似概念を回答させる（知識体系化）ことも可能です．
知識深化，知識体系化の二つの方向を組み合わせてLLMから知識を得て全体像を構成します．

### 情報補完法

ユーザーが説明のために多くの文章を書くのは大変なので，入力は必要最低限にし，LLMに文章を補完させます．
学校の授業を参考にすると，教師が一般的な知識を提供し，生徒が質問する流れが使えます．
学校の授業においては，教師がまず一般的な知識の提供を行い，生徒は質問を行います．
短い質問でも共通認識があるため，妥当な回答が得られるでしょう．
先生役をLLM，生徒がユーザーとしてプロンプト作成にもこの方法を用います．
また，学校では一度にすべてを教えません．
例えば，数学の場合，小学校の算数，中学校の数学，高校の数学，大学の数学と徐々に難しくなるだけでなく，詳細を学ぶようになります．
商用LLMには回数制限があるので，一度で多くの回答を求めがちですが，すべての回答を一度に求めるのではなく，共通認識を基に細部を知識として得るようにします．

### プロンプトフォーマット

野口竜司氏の「文系AI式7Rプロンプト」\cite{RefNoguchiLLMprompt}を参考にしています．
提唱されている全ての要素を取り入れませんが，質問を明確に分けて記述することは思考の整理になるとともに，プロンプトにおける'変数'の利用法も参考になります．
質問項目は定形文を予め作り，そこから選ぶことで必須項目を明確にできます．

## LLMの限界とユーザーの責任

LLMは，膨大なデータセットを基に学習し，人間が使う言語や科学技術の記述を文脈も含めてベクトル空間に写像し，その空間内で計算を行うことで多様な質問に答える能力を持っています．
しかし，これらのモデルは時として錯誤な振る舞いを示し，「ハルシネーション」と呼ばれる現象を起こし，誤った内容を生成することがあります．
このような誤情報は，特に正確な情報が必要とされる状況で問題となるため，ユーザーはLLMの回答を鵜呑みにせず，批判的に評価し，常に検証することが重要です．

また，ハルシネーション以外にも予期しない回答を得ることがあります．
これは質問者が，LLMが会話の文脈をベクトル空間に写像し，適切な計算を行って返答を出力する装置であることを意識しないことから起こると思われます．
簡単な質問に対して妥当な回答を返すことから，より複雑な質問に対しても，文脈などを意図的に提供せずとも，LLMは簡単な質問と同様に妥当な回答をする能力があると勘違いして期待してしまいがちです．
人間同士の会話における意思疎通を図ること過程を考えてみましょう．会話相手が理解できないことを即座に質問し返す反復過程を経り，互いに自分の知識を提供する双方向のプロセスを通じて意思疎通を図ります．一方，LLMへの質問では，意図を言語化する過程を省略しがちです．
読者自身でLLMに質問して得られた回答の正確性に自信が持てるのであればハルシネーションが起きても正確な回答だけ選択できるのですが，そういう方は既に初心者ではありません．

現状では，ユーザーは以下のことが求められます：

1. 最低限の知識を持ち，
1. 明確な質問をして，
1. LLMの回答を評価し，取捨選択する，

\noindent
そして，データ解析学手法の知識があればあるほど，的確にそして高速にこれらを行うことができます．

現状では，LLMはあくまで知識獲得や問題解決の支援を行うものであり，その回答をユーザーが自分で再構築する必要があります．更に，LLMはこの再構築も客観的にサポートできます．
再度申し上げますが，ユーザーは「LLMから得られる回答を鵜呑みにせず，批判的に評価し，結果を常に検証する」ことが重要です．
