# データベースの正規化

（リレーショナル）データベースには
データの冗長性を排除し、更新時の整合性を維持しやすくすることを目的とした、第N規格化という考え方があります。

主キー(データベースを一意に認識するための項目。これを用いるとデータベースはデータに高速にアクセスできるように作られる。）を定義した後に

1. 第１正規形：1つのセルには1つの値しか含まれなず、各繰り返し部分に1つずつ独立したレコードを作り、横方向の長さを統一されているデータベース。
2. 第２正規形：関数従属とは主キーが決まると、列の値が一意に定まる関係である。第１正規形から部分関数従属を排除したデータベース。
3. 第３正規形：主キー以外の列に関数従属している列のことを推移的関数従属と呼び、第２正規形のテーブルから、推移的関数従属している列を分離したデータべース。



遷移金属元素(TM)と希土類元素(RE)からなる物質を考えます。
ここで物質というのは例えば、CuCo5構造SmCo5が具体的に指す結晶構造のことで、主キーとして「物質」要素があります。

## 第１正規形
第１正規形は以下の表です。

|物質| 構造名 |　TM元素名| TM特徴1| TM特徴2  | RE元素名|  RE特徴1| RE特徴2  | TM/RE組成比|
|---|---|---|---|---|---|---|----|---|
|CuCo5構造SmCo5 |CuCo5構造 | Co| Z(Co)| S3d(Co) | Sm|  Z(Sm)  | (g-1)J4f(Sm) | 5/1 |
|CuCo5構造MdFe5 | CuCo5構造|Fe| Z(Fe)| S3d(Fe) | Nd|  Z(Nd)  | (g-1)J4f(Nd) | 5/1 |
| Zn17Th2構造SmCo17 |Zn17Th2構造|Co| Z(Co)| S3d(Co) |Sm|  Z(Sm)  | (g-1)J4f(Nd) | 17/2|
|Zn17Th2構造Nd2Co17 | Zn17Th2構造|  Co| Z(Co)| S3d(Co) | Nd| Z(Nd)  | (g-1)J4f(Nd) | 17/2 |

機械学習を行うにはこの形にすると便利です。

一般にデータベースをつくる際は、セルの間に従属関係があると一つのキー（例えばTM特徴２）を変化させると、表全ての値を変化させないといけません。このために高次の正規化を行います。

## 第２正規形
主キーである物質が決まると構造名、TM元素名、RE元素名が決まり、
構造名、TM元素名、RE元素名からそれぞれTM/RE組成比、TM特徴1とTM特徴2 、RE特徴１とRE特徴２が決まります。


第二正規化を行うと主キーから
物質名を決めると定まる化学式、TM元素名、RE元素名を切り離します。

|物質名| 構造名 | TM/RE組成比|
|---|---|---|
|CuCo5構造SmCo5 |CuCo5構造| 5/1 |
|CuCo5構造NdFe5 |CuCo5構造| 5/1 |
| Zn17Th2構造SmCo17 | Zn17Th2構造 | 17/2|
|Zn17Th2構造Nd2Co17 | Zn17Th2構造| 17/2 |


|物質|　TM元素名| TM特徴1| TM特徴2  |
|---|---|---|---|
|CuCo5構造SmCo5 | Co| Z(Co)| S3d(Co) |
|CuCo5構造NdFe5 |Fe| Z(Fe)| S3d(Fe) ||
| Zn17Th2構造SmCo17 |Co| Z(Co)| S3d(Co) |
|Zn17Th2構造Nd2Co17 |   Co| Z(Co)| S3d(Co) | 

|物質|  RE元素名|  RE特徴1| RE特徴2  | 
|---|---|---|---|
|CuCo5構造SmCo5 | Sm|  Z(Sm)  | (g-1)J4f(Sm) | 
|CuCo5構造NdFe5 | Nd|  Z(Nd)  | (g-1)J4f(Nd) |
| Zn17Th2構造SmCo17 |Sm|  Z(Sm)  | (g-1)J4f(Nd) | 
|Zn17Th2構造Nd2Co17 | Nd| Z(Nd)  | (g-1)J4f(Nd) | 

それぞれの表に重複があり、更に整理することが可能です。


## 第３正規形

主キー以外で、
それぞれの表にある推移的関数従属（例えば、A->B, B->Cで、A->Cの間の間接的な従属）を更に切り離します。

|物質| 構造名 | 
|---|---|
|CuCo5構造SmCo5 |CuCo5構造|
|CuCo5構造NdFe5 |CuCo5構造|
| Zn17Th2構造SmCo17 | Zn17Th2構造 |
|Zn17Th2構造Nd2Co17 | Zn17Th2構造|



| 構造名 | TM/RE組成比|
|---|---|
|CuCo5構造| 5/1 |
|Zn17Th2構造 | 17/2 |



|物質|　TM元素名|
|---|---| 
|CuCo5構造SmCo5 | Co| 
|CuCo5構造NdFe5 |Fe|  
| Zn17Th2構造SmCo17 |Co| 
|Zn17Th2構造Nd2Co17 |   Co| 

|　TM元素名| TM特徴1| TM特徴2  |
|---|---|---|
| Co| Z(Co)| S3d(Co) |
|Fe| Z(Fe)| S3d(Fe) |

|物質|  RE元素名| 
|---|---| 
|CuCo5構造SmCo5 | Sm| 
|CuCo5構造NdFe5 | Nd| 
| Zn17Th2構造SmCo17 |Sm| 
|Zn17Th2構造Nd2Co17 | Nd| 


|  RE元素名|  RE特徴1| RE特徴2  | 
|---|---|---|
| Sm|  Z(Sm)  | (g-1)J4f(Sm) | 
| Nd|  Z(Nd)  | (g-1)J4f(Nd) |




第３正規形でデータベースを作成しておけば、TM特徴２をS3dから別の特徴量に変えた場合に、参照は増えますが正しく第１正規形の形を作り直せることが分かると思います。

説明変数を生成する際は頭の中でこれを行っているので言うまでも無い（自明）と思うかも知れませんが、
まず、このような考え方自体を明確に概念化するのは自明ではありませんし、
更に、今の例で、RE/TM比を構造名の関数従属関係としたのも自明ではありません。

この作り方（物質ー＞構造名ー＞TM/RE比）だとドープした場合に対応できません。構造から直接TM/RE比を生成した方が良いかもしれません。
