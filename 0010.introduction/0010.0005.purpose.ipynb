University at Buffalo, Department of Materials Design and Innovationでは
学生が

## 大規模言語モデルを利用した学習

### 新しいツールとしての大規模言語モデル

大規模言語モデル（Large Language Models、LLM）の進化は、教育を劇的に変化させます。LLMは、特に言語学習やプログラミング教育において顕著な効果的と思います。
例えば、プログラミング学習において、LLMを利用することで、初心者でもPythonなどの言語の基本を理解しやすくなります。また、LLMは、簡単なコード例を提供したり、より複雑なプログラムの作成をアシストしたりすることが可能です。生徒が「回帰モデル作成のコードを書きたい」と質問すると、LLMは具体的なコードの例とともに、その手順を分かりやすく説明します。

従来の学習方法は主に講義や教科書といった静的なメディアに依存していました。これらの方法では情報の流れが一方向で、生徒から教師へのフィードバックや質問の機会が限られていました。しかし、LLMの出現により、学習プロセスはよりダイナミックでインタラクティブになります。

たとえば、生徒がデータ分析に関する質問を持っている場合、LLMはその疑問に応じて具体的な例やリソースを提供できます。これにより、生徒は自分のペースで学び、理解を深めることができます。さらに、LLMは学習者の質問に基づいて、具体例に即した返答や解説を行うことができます。これは従来の教科書や講義では不可能でした。

また、LLMを利用することで、生徒は自分の学習に積極的に関与することが可能になります。例えば、生徒が機械学習のアルゴリズムについて質問した場合、LLMはそのトピックに関連する概念を説明し、実際のデータセットで試すためのサンプルコードを提供することができます。これにより、生徒は理論だけでなく、実際のアプリケーションにも触れることができ、理解を深めることができます。

### TeachingからFacilitatingへ

Teachingは教授ですが、
Facilitatingは聞き慣れない言葉と思います。後者はコーチングなどと言われることもある言葉と思います。

LLMは、単に新しいツールの導入にとどまらず、教育方法そのものにも影響を与えるでしょう。
LLMの活用により、教育が「教える（Teaching）」から「促進する（Facilitating）」への移行することが期待されます。

伝統的な教育は、教師が知識を伝え、生徒が受け取るという受動的な形式でした。
しかし、LLMを取り入れた新しい形式では、生徒自身がより能動的な役割を担うことが可能になります。
生徒は自分の目的に基づいて学習を進め、具体的な問題に即した、問題解決方法を教授されることになります。近い将来にこれが可能となるでしょう。


しかし、この方法には留意点もあります。特に初心者にとっては、より構造化された指導が必要な場合があります。また、基礎的な概念の理解が不十分なまま先に進むと、誤解や学習のギャップが生じる可能性もあります。また、現在はLLMの回答は必ずしも適切ではないので、LLMの回答する誤った理解のまま学習してしまう可能性もあります。そのため、特に最初のうちは一度「正しい」概念、用語を説明された後にLLMの使用を行った方が好ましいでしょう。

### LLMの回答の捉え方

本資料はその順序で行いますが、今後読者が新規概念を自ら学習する場合は、LLMの回答が正しいのか判断できるだけの知識を持つことも必要であるということにも留意してください。
例えば、AI翻訳は一般的なものですが、AI翻訳が適切でない回答を示すことに気づかれる方は多いと思います。文脈により適切な翻訳は変わりますので、翻訳結果が正しいかどうかを選択するだけの知識がないとAI翻訳を利用しても適切な翻訳結果を得ることはできません。

また、大きな科学的発見は既存の概念を壊すものであることが多く、LLMの出力は変わることがない固定した知識ではなく、知識例であることも留意してください。
例えば、相対性理論はニュートン力学の概念を覆すものでした。
また例えば、1972年にMottは金属絶縁体転移では金属側に最小電気伝導度があると主張し、これらの研究で1977年にノーベル賞を受賞しました。
一方、1978年発表のAbrahams、Anderson、 Licciardello 、 Ramakrishnanによるアンダーソン局在のスケーリング理論では三次元の金属絶縁体転移では絶対零度の電気伝導度は０から連続的に変化（つまり金属絶縁体転移は二次転移）し、これは最小電気伝導度の概念を覆すものでした。

先行研究のみを学習したLLMでは後発研究の結果は出ないと思われます。
web知識の総和であるLLMの回答は完成された知識ではなくその時代で常に発展中の知識や理論の状況を反映した現在進行形の知識の一部、
問題解決のためには有用であると理解しつつもそれに疑いを持つという矛盾した姿勢で捉える知識であると考えます。
もし、それが演繹的な科学知識である場合は、教科書を別途参照し、どう捉えるべき知識かを確認してください。
