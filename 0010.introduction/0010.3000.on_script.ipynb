# jupyter notebookで記述する利点

1. memory上に変数が置かれます。データのロード・セーブなしに色々と実行が可能です。
1. jupyter notebookではテキスト出力だけでなく、図を同時に記載することが出来るます。先生に見せる場合に、どの順序で何をしたのかという解析過程を同時に見せることが可能です。また、解析の誤り（、そして研究捏造）が疑われることがあった場合に、解析順序をふくめて結果見せることができます。


# scriptについて

このscriptは単一programとして上から順に実行することを想定しています。
複数回実行できるcellもありますが、
途中で任意の同じcellを数回実行することは想定していません。
