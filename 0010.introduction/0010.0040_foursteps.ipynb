# 機械学習の四過程

前章では，予測モデルの学習・選択過程について説明した．そして、一般的に物理現象を機械学習過程に落とし込むには付随する過程が存在することも説明した．
これをまとめると、一般的に、機械学習過程は以下のデータ収集，データ加工，データからの学習，可視化の
四つの過程に分けて考えることができる．


前章で説明した結晶構造から全エネルギーを予測する問題は
機械学習古典ポテンシャルを学習する問題であり、多くの論文がある。
この問題のそれぞれの過程は以下のように分けられる。

1. データ収集：結晶構造（原子種類と原子座標の順序リスト）を
用意して第一原理計算を行い，それぞれの構造の全エネルギー
を計算を得る過程である．

2. データ加工：観測結晶データ（「観測生データ」と呼んでおく。）である原子種類と原子座標の順序リストは
同種原子の順序変換に関しても等価でなく，座標系を決めた原子座標は並
進，回転に対して等価でない．このため，これらの対称操作に関して不変
になるように変換し，「加工済み観測データ」を生成することが多い．
前章で導入したRDFは加工済み観測データである。
更に，
一般的な機械学習手法ライブラリを用いるために単位，スケールが異なる
説明変数の「データ規格化」を行う。

3.  データからの学習：ハイパーパラメータがあれば交差検定などでハイパー
パラメータの値を決め，回帰モデルを学習する．

4. 可視化（随時実行）：大量のデータを扱うため人間が途中経過や結果を理解
できるように可視する過程である．
