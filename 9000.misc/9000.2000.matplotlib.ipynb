## matplotlibの簡単な説明

matplotlibはpythonの標準可視化ライブラリです。

可視化するデータの作成をします。

```python
import numpy as np
x = np.linspace(-1, 1, 20)
y = x**2
y2 = np.sin(x)
```

jupyter notebook上でmatplotlibで図示をするためには
以下をまず記載します。

```python
import matplotlib.pyplot as plt
%matplotlib inline
```

最近は%matplotlib inlineが無くても表示できる場合が多いようです。

plot()は標準では線でつなぎます。

対数表示を行う。

```python
x = np.logspace(-5,1,10)
y = np.exp(x)

plt.figure()
plt.xscale("log")
plt.yscale("log")
plt.plot(x,y,label="data")
#plt.loglog(x,y)
plt.xlabel("x")
plt.ylabel("y")
plt.title("x-y")
plt.legend()
plt.show()
```

loglog()以外にもsemilogx(), semilogy()があります。

ファイル保存だけして、図示しない場合。

```python
import os
os.makedirs("image_executed", exist_ok=True)
plt.figure()
plt.hist(y,bins=10)
plt.savefig("image_executed/loglogplot.png")
plt.close()
```

## 複数plot

```python
# データを作り直す。
x = np.linspace(-1, 1, 20)
y = x**2
y2 = np.sin(x)
```

```python
# 縦

fig, axes = plt.subplots(2,1) #plt.subplots(1,2)だと横に並べる。

ax = axes[0]
ax.plot(x,y,label="obs")
ax.set_xlabel("x")
ax.set_ylabel("y")
ax.set_title("x-y")
ax.legend()

ax = axes[1]
ax.plot(x,y2, label="pred")
ax.set_xlabel("x")
ax.set_ylabel("y")
ax.set_title("x-y2")
ax.legend()

fig.tight_layout()
fig.savefig("image_executed/xy.png")
```
