# PDF 文書を使った検索拡張生成（RAG）

検索拡張生成（Retrieval-Augmented Generation, RAG）では、質問に関連する文書の一部を検索し、その内容を根拠に大規模言語モデル（LLM）が回答を生成します。この章では ARiSE と SPReAD の公募要領 PDF を使います。



## 実習の準備

実習に使う Python 環境を有効にし、ターミナルで `8300.LLM/` に移動して [依存パッケージ](requirements.txt) をインストールします。

```bash
python -m pip install -r requirements.txt
```

Jupyter を起動するターミナルで `OPENAI_API_KEY` を設定します。`YOUR_API_KEY` は自分のキーに置き換えてください。

Linux / macOS

```bash
export OPENAI_API_KEY="YOUR_API_KEY"
```

Windows PowerShell

```powershell
$env:OPENAI_API_KEY="YOUR_API_KEY"
```

同じ環境から JupyterLab を起動し、実習に使う Python カーネルを選びます。


## 1. PDF を登録する

PDF から抽出したテキストを短い断片（チャンク）に分け、OpenAI Embeddings でベクトルに変換します。ベクトルと出典情報を Chroma に保存します。



## 2. 文書を検索して回答を生成する

質問をベクトルに変換し、Chroma で関連するチャンクを検索します。検索結果を質問とともに LLM に渡して回答を生成し、参照した出典を確認します。
