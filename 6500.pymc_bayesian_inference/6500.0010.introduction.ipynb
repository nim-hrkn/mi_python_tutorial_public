## 解説

関数モデル$f(x)$が複雑になると下図のように最小値が複数現れます。
![gaussian fluctuation](image_keep/fig_fluctuation_coefficient2.PNG)

平均場からのゆらぎを考える場合と同じく、どこから係数のゆらぎを展開するのか指定が必要です。


線形ベイズ回帰では
priorの指定により最小値$c^{min}$とそこからの揺らぎ（共分散行列）も異なる解になり、またそれらの計算が解析的に可能です。


ベイズ回帰においては揺らぎの展開点をベイズ推定のpriorとして指定することになります。
例えば、距離に依存するパラメタを持つ古典スピンモデルを全エネルギーに対して回帰する場合を考えます。

$$
 H = \sum_{i,j} J_{i,j}(r_{i,j}) S_i S_j
$$

のパラメタ$J_{i,j}(r)$を求める場合には遠距離になるほど$|J_{i,j}(r)|$が
小さいという物理的制約をpriorで与えて最適係数を求めても良いでしょう。

![posterior](image_keep/posterior.PNG)


目的変数$y(x,w)$は基底関数$x$と係数$w$の線形方程式で書けるとします。
$$
y(x) = x^T w
$$
観測値にはyに更にノイズが入るとし、
y(x,w)とガウスノイズから目的変数$t$が得られているとします。
$$
t = N(y(x,w)|\beta)
$$

観測値がある値ベクトルtを取る確率は
は$t$,$x$の各成分$t_i$, $x_i$を用いて

式（１）:

$$
P(t|x,w,S^{-1} )
= \exp( -\frac{1}{2} \sum_i (t_i - x_i^T w)^T \beta_i^{-1} (t_i - x_i^T w ) ) 
$$
ともかけます。今は$\beta_i=\beta$です。
以下
では$x$,$\beta$を顕に書くのをやめてこれを $P(t|w)$書いておきます。


問題は
tが与えられた時にwを求めることです。
$P(t|w)$から$P(w|t)$への変換はベイズの定理を用いて
$$
P(w|t) = P(t|w) P(w) /P(t)
$$
と書けます。
$P(t)$ は$P(w|t)$を計算する上では定数なので$P(t)=1$としておく。
つまり
$$
P(w|t) \propto P(t|w) P(w)
$$
です。
$P(w)$の部分をprior, $P(t|w)$の部分をlikelihood,$P(w|t)$をposteriorと呼びます。
