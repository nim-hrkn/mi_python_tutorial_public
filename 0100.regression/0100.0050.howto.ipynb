## scikit-learnクラスの説明

以下では説明変数をX、目的変数をyとします。
規格化されていない説明変数をXrawと置きます。

この章で紹介するscikit-learnクラスの使い方をまとめます。


Xraw: (N,P)のarray.　規格化していないしていない説明変数。

X: (N,P)のarray. 規格化済み説明変数。

y: (N)のarray. 目的変数。

### 規格化

* 規格化(StandardScaler)

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
scaler.fit(Xraw)
X = scaler.transform(Xraw)
```

### データ分割

* 訓練データとテストデータの分割

```python
from sklearn.model_selection import train_test_split
Xtrain, Xtest, ytrain, ytest = train_test_split(X,y,test_size=0.25) 
```

* 10回交差検定

```python
from sklearn.model_selection import KFold
kf = KFold(n_splits=10,shuffle=True)
for train_index, test_index in kf.split(X):
        Xtrain = X[train_index]
        Xtest = X[test_index]
        ytrain = y[train_index]
        ytest = y[test_index]
```

### 線形回帰

r2_scoreは他にも計算クラスがあります。

```
from sklearn.linear_model import LinearRegression
reg = LinearRegression()
reg.fit(Xtrain,ytrain)
ytestp = reg.predict(Xtest)
r2_score = reg.score(Xtest,ytest)
```

* ハイパーパラメタalphaを与えた

### ハイパーパラメタの自動決定を行う回帰関数

#### 線形回帰の場合

* 10回交差検定を行うLassoCVの利用方法

```python
from sklearn.linear_model import LassoCV
from sklearn.model_selection import KFold
kf = KFold(n_splits=10,shuffle=True)
reg = LassoCV(cv=kf)
reg.fit(X,y)
```

10回交差検定を用いて最適なハイパーパラメタを決定します。
最後に(X,y)を用いて一つのRidge回帰モデルを生成します。


* 10回交差検定を行うRidgeCVの利用方法
```python
from sklearn.linear_model import RidgeCV
from sklearn.model_selection import KFold
kf = KFold(n_splits=10,shuffle=True)
reg = RidgeCV(cv=kf)
reg.fit(X,y)
```

* 回帰オブジェクトregを用いた予測と$R^2$の計算法

```python
yp = reg.predict(X)
r2_score = reg.score(X,y)
```

#### KernelRidge回帰の場合

* GridSearchCVによるRBFカーネルを用いたKernelRidgeのハイパーパラメタ最適化

```python
from sklearn.model_selection import KFold
from sklearn.kernel_ridge import KernelRidge
from sklearn.model_selection import GridSearchCV
kf = KFold(10,shuffle=True)
estimator = KernelRidge(alpha=1,gamma=1,kernel="rbf")
reg = GridSearchCV( estimator,
          cv=kf, param_grid={"alpha":np.logspace(-6,0,12),"gamma":np.logspace(-4,0,13)})
reg.fit(X,y)
```
