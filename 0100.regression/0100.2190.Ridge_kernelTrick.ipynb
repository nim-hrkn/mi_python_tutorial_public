###  Ridge回帰とKernel Ridge回帰

Kernel='linear'によりカーネル回帰は線形回帰と同じになります。

（カーネルトリック）

```python
import warnings
import sys
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
%matplotlib inline

pd.set_option("display.max_rows", 20)
pd.set_option("display.max_columns", 80)

# warning error を表示しない。
warnings.filterwarnings('ignore')
```

```python
df = pd.read_csv( "../data_calculated/x5_sin.csv")
descriptor_names = ['x1', 'x2', 'x3', 'x4', 'x5', 'x6']
target_name= 'y'
Xraw = df.loc[:, descriptor_names]
y = df.loc[:, target_name]
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
scaler.fit(Xraw)
X = scaler.transform(Xraw)
```

切片項を説明変数の方に加えます。

```python
one = np.ones(X.shape[0]).reshape(-1, 1)
X1 = np.hstack([X,one])

# sklearnのRidgeの内部処理がyの平均を引くので、同じ処理にするために平均を引いておきます。
y1 = y - np.mean(y)
```

```python
from sklearn.kernel_ridge import KernelRidge
from sklearn.linear_model import Ridge
score_list = []
score_kr_list = []
for alpha in np.logspace(-3, 3, 10):
    reg = Ridge(alpha=alpha, fit_intercept=False)
    reg.fit(X1, y1)
    yp = reg.predict(X1)
    score = reg.score(X1, y1)

    reg_kr = KernelRidge(alpha=alpha, kernel="linear")
    reg_kr.fit(X1, y1)
    yp = reg_kr.predict(X1)
    score_kr = reg_kr.score(X1, y1)

    print("{:10.3f} {:.3f} {:.3f} {:.9E}".format(
        alpha, score, score_kr, score-score_kr))
    score_list.append(score)
    score_kr_list.append(score_kr)
```
