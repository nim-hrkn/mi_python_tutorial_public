## 例題１

Q. 解明しようしている材料問題（仮想問題）が何か？

A. 適切な前処理を行うことで説明変数sin(x)が選択されることを確認する。

Q. データがどうようなものか？

A. 説明変数 : x,x^2,x^3,...,sin(x)
A. 目的変数 : sin(x)


Q. どのような手法で何を抽出しようとするか？

A. 線形回帰、LASSO, Ridge回帰を用いて訓練・テストデータ分離による性能評価、クロスバリデーションによる予測性能評価を行う。
