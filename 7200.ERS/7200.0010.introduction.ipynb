## 解説

用語の使い方がまた異なります。
「予測」ではなく「推論」や「推薦」を用います。

### 証拠を積み重ねて結果を推論する。

誤り、欠損、データインスタンス数の違い、を考慮する手法の紹介をします。

### wikipediaのDempster–Shafer理論の紹介

https://wikipredia.net/ja/Dempster%E2%80%93Shafer_theory

「直感に反する結果を生み出す可能性があります。」の例を見て、否定する方が多いようです。
おそらく学生はこれを見ると証拠理論を勉強、実践しようというやる気を無くすでしょう。
しかし、上の例のような病的な場合を考えなければDempster–Shafer理論を用いた推論は有用です。

### 簡単な例
幾つかの元素組み合わせである構造ができるかの実験を行いました。
物質が壊れやすく構造を測定するのが難しい物質とします。誤りもあり実験者により結果が異なるでしょう。

| 実験者名 |構造名 | 存在 |
|---|---|---|
| A | 構造1 | 有 |
| B | 構造1 | 無 |
| C | 構造1 | 無 |
| D | 構造1 | 有 |
| E | 構造1 | 有 |
| A | 構造2 | 有 |

物質に対してまとめます。各物質が作成できるかを平均割合で確率として定義すると、
| 物質 | 有 | 無| 
|---|---|---|
物質1|3|2| 
物質2|1|0| 

です。確率に直します。
各物質が存在する確率は

$$ P(物質1) = 3/5 $$,
$$ P(物質2)= 1/1 $$

$ P(物質2) > P(物質1) $つまり、物質2の方ができやすいという結論になります。

各物質が作成できるかを
有を優先する。つまり、

| 物質 | 有無| 
|---|---|
物質1|有|
物質2|有| 

無を優先する。

| 物質 | 有無| 
|---|---|
物質1|無|
物質2|有| 

もあります。どの選び方を採用するかは、用途により選択は異なる、とも言えます。

解析では、
確率論の拡張として
**物質１では物質２より実験数が多い、そして物質１では有の数が多い、いう知見**を取り入れたいところです。


### 固溶体の例

固溶体：結晶。長距離秩序はあるが、原子位置に各元素はランダムに存在する物質

固溶体 vs 液体 の自由エネルギーの比較で固溶体が存在しない元素組み合わせもあります。
この結果の抜粋が以下に表のように成っているとします。

| 物質 | 存在 |
|---|---|
| Cr, Fe|有|
|Cr, Nb |無|
|Cr, Mn, Tc |有|
|Cr, Ni, Pd |有|
| Co, Cr, Ni, Pd | 有|
| Ag, Cr, Ni, Pd | 無|
|...|...|

元素Crは多くの実験があります。そして元素の組み合わせ例も有り、無しが多くあります。
Crに対して妥当な推論（予測）ができるかもしれない。

一方、元素Tcは一つのデータ（有）しかありません。
データベースを全部探してもごく少ないデータしかないとします。
しかし、その中でも、元素Tcが含まれる物質は存在する、とします。

この場合に元素Tcが含まれる物質は存在する、という結論を出したいでしょうか。
むしろ、
データ（証拠）が少なくて推論できない、という結論を導いたほうが良いでしょう。
そしてデータインスタンス数が多いCrとは推論の確からしさが大きくことなるはずです。
どの程度有り、無し、と推論できるのか、有り、無しと推論できないのかを定量的に評価する手法が証拠理論になります。


### 証拠理論の枠組み

固溶体の例で物質がある有(A)、無(B)とます．
このA,Bを核集合と言います．
AかBか分からないは「有または無((A,B})」です。
そして、これらを集合として書くと
$ \{\{A\}, \{B\}, \{A,B\}\} $です。
数学的な枠組みに載せるために空集合($\emptyset$)も入れておき、
各要素が、
$$ \{\emptyset, \{A\},\{B\}, \{A,B\}\} $$
とします。
それぞれに対して、割り当て値が$m$あると定義します。
これを**信念割当関数**（マス関数）と言い，各mを**確信度**と呼びます。
インデックス$P$のデータインスタンスについて
$$ \vec{m}_P = ( m_P({\emptyset}), m_P(A), m_P(B), m_P(\{A,B\}))$$
です。
更に、確率のように扱うために、
$$m(i)\ge 0$$
$$m(\emptyset) = 0$$
$$ \sum_i m(i)=1$$
と定義します。

観測データがデータインスタンス、インデックス：P, Q, R,...持つとして、
$$\{\vec{m}_P , \vec{m}_Q,\vec{m}_R,...\}$$
を持ちます。これを**信念構造**（＝信念割当関数の集合）と定義します。




固溶体を持つの例では、
- HEA（固溶体がある）＝A, 
- $\neg$HEA（固溶体が無い）＝Bとします。
- 観測がない（＝証拠が無い）、は$\emptyset$ではなく、{HEA, $\neg$HEA}です。


## 信念の結合

ある物質に実験P（観測P）と実験Q（観測Q)があるとします。
その物質の
観測Pの信念割当関数$\vec m_P$と
観測Q信念割当関数$\vec m_Q$と
の２つの信念構造を情報結合します。統合（結合）後の信念割当関数を$\vec m_P \oplus \vec m_Q$と書くことにします。

この過程は

1. 論理積を行う
1. 信念割当関数の値の定義に合うように変換する。

で達成されます。

行（縦）がP, 列(横）がQとします。
$ \{\emptyset, {A}, {B}, \{A,B\}\} $の論理積は積以下になります。

| * | $\emptyset$ | {A} | {B} | {A,B} |
|--|--|--|--|--|
|$\emptyset$| $\emptyset$ | $\emptyset$ | $\emptyset$ | $\emptyset$ |
|{A}| $\emptyset$ | {A} | 0 | {A} |
|{B} | $\emptyset$ | 0 | {B} | {B } |
{{A,B} | $\emptyset$ | {A}| {B}| {A,B} |

単純な論理積とすると場合の統合後の信念割当関数は$ m_P(i) m_Q(j)$となります。ここでi,jは
$ \{\emptyset, \{A\},\{B\}, \{A,B\}\} $です。

まず、{A} and {B}は$\emptyset$です。
しかし、$m_P({A})$が非０，$m_Q({B})$が非０の場合は$m_P({A}) m_Q({B})$が有限の値を持ちますが、信念割当関数の定義から$\vec m_P \oplus \vec m_Q(\emptyset)=0 $とせねばなりません。
最も簡単な**Dempsterの規則**ではでは単純に$\emptyset$となる要素積を捨てて、規格化し直します。
上の表の場合は具体的には、$\emptyset$となる要素積の$ m_P({A}) m_Q({B})+m_P({B}) m_Q({A})$を$K$とし、
そして$1/(1-K)$でかけて$\vec m_P \oplus \vec m_Q(i)=1$ となるように規格化します。

また、$\vec m_P \oplus \vec m_Q = \vec m_Q \oplus \vec m_P$ですので結合が交換則を満たします。
データ更新が有る場合に、交換則を満たすと証拠統合結果が順序によらず値が一意に決まるので交換則を満たすことは重要です。

## 信念割当関数の値の割当

$\vec m = (\emptyset, m(\{有\}）, m(\{無\}）,m(\{有,無\})$とします。
観測PとQとRを行い
- 観測Pで物質が存在する場合は $\vec m_P= (0, 1, 0, 0 )$
- 観測Qが存在しない場合は $\vec m_Q = ( 0, 0, 1, 0 )$
- (実験が無い場合は
$\vec m_R = ( 0, 0, 0, 1 )$)

が一つの決め方です。(後で説明しますが、「実験が無い場合」は使いません。)

観測が難しく、誤りの可能性があるとすると実際の適用では境界をぼかします。
その例は、
- $\vec m_P = ( 0, a, 0, 1-a )$
- $\vec m_Q = ( 0, 0, a, 1-a )$
- ($\vec m_R = ( 0, 0, 0, 1 )$)

です。
データの質が同じとすると、全てのデータインスタンスで同じ$a$です。
（ここで、ある実験が「完全に正しい」のであれば、その信念割当関数では$a=1$としても良いでしょう。）
他の決め方ももちろんあります。
この場合も新帰納法ですので、ハイパーパラメタとして妥当な推論ができるかで決めます。

### 信念割当関数結合の計算例

具体的に計算してみます。
観測P、観測Qである物質の存在がともに有の場合の信念割当関数を結合します。
- $\vec m_P= ( 0, a, 0, 1-a )$
- $\vec m_Q= ( 0, a, 0, 1-a )$

とすると
$$\vec m_P \oplus \vec m_Q = ( 0, 2a-a^2, 0, (1-a)^2)$$

です。証拠が$n$個あると、統合後の$m(\{有,無\})= (1-a)^n$で減少していきます。確信度の和が1なので$m(\{有\})$がその分大きくなることが分かります。

一方、
観測Pで有、観測Qで無という矛盾した観測がある場合の信念割当関数を結合します。
- $\vec m_P= ( 0, a, 0, 1-a )$
- $\vec m_R= ( 0, 0, a, 1-a )$
この場合
$$ \vec m_P \oplus \vec m_R = (0, a/(a+1), a/(a+1), (1-a)/(1+a))$$

（本は最後の部分が$(1+a)/(1-a)$となっていて誤りです。正しくは$(1-a)/(1+a)$です。)
$a>0$なので、統合後の$\vec m_P \oplus \vec m_R(\{有\})$、統合前の$m_P(\{有\})$より小さくなっています。
また、
$$ \vec m_P \oplus \vec m_R(\{有,無\}) - m_P(\{有,無\}) = - a(1-a)/(1+a)$$

なので、統合後の{有,無}=分からないの確信度は**証拠が増えたので**減っています。
このように証拠に応じてそれぞれの確信度が変化していきます。


## ERS scriptの実行過程

### 目的と観測データ

**目的は存在・非存在の元素等比固溶体の推論**です。

そのために以下の論文の元素等比固溶体観測データを用います。
観測データでは３，４，５元元素組み合わせの元素等比固溶体が、存在する、しないがまとめられています。

Ref.
" Y. Lederer, C. Toher, K. S. Vecchio, and S. Curtarolo, “The search for high entropy alloys: A high-
throughput ab-initio approach,” Acta Materialia 159, 364 (2018)."

このデータは実験、T=0の理論、有限温度の理論計算結果を含みます。
T=0の理論データを用い、観測データに誤りの可能性があるとして証拠理論を用いて推論を行います。


### 推論過程

２，３，４，５元元素組み合わせの元素等比固溶体があります。
２，３，４，５元系と大きくなるに連れ、理論的な可能な元素組み合わせ数が爆発的に多くなるのは理解できると思います。(N_C_m)
一方、理論的な可能な元素組み合わせ数に対して観測データ比率は５，４，３，元系と大きくなります。
５元系ではごく僅かな比率しか観測データが存在しません。
って、５元系の予測を行うには、２，３，４元型の観測データから何らかの知見を定量的に得る必要があります。

これを実現するために、類推過程は直接推薦するのではなく「置換」を主として考え、以下の過程を採用しました。

1. 観測データから元素組み合わせ置換の類似度の信念割当関数求めます。
2. 存在する・しない固溶体から元素組み合わせを置換した場合の固溶体の信念割当関数を求めます。

過程１，２どちらにも複数の証拠がありますから、証拠理論によりそれらを統合（結合）します。

#### 1. 観測データから元素組み合わせ置換の類似度の信念割当関数

見にくいかもしれませんが、論文の表記に合わせます。

Ref. "Minh-Quyet Ha, Duong-Nguyen Nguyen, Viet-Cuong Nguyen, Hiori Kino, Takashi Miyake, and Hieu-
Chi Dam, "Evidence-based recommender system for high-entropy alloys", Nature Computational Sci-
ence, 1, 470 (2021)."

物質$A_i$で、元素組み合わせ$C_t$を元素組み合わせ$C_v$に変えると物質$A_j$になるとします。
つまり、ある元素組み合わせをXとして
- 物質$A_i$は{ X, $C_t$} 
- 物質$A_j$は{ X, $C_v$} 

と共通のXがあるということです。

$A_i$と$A_j$のどちらも固溶体が存在する・しない場合$A_i$と$A_j$は似ているとします。
この信念割当関数を
$$\vec m = ( m(\emptyset)), m(\{類似\}), m(\{非類似\}), m(\{類似,非類似\}))$$
に対して、
$$\vec m_{A_i, A_j}^{C_t,C_v} = (0, a, 0, 1-a)$$
とします。

一方、
$A_i$と$A_j$のどちらかは固溶体が存在する、もう片方は存在しない場合$A_i$と$A_j$は似ていないとします。
つまり、
$$\vec m_{A_i, A_j}^{C_t,C_v} = (0, 0, a, 1-a)$$
とします。

$A_i$と$A_j$の固溶体データが存在しない場合もあります。その場合は「情報・証拠の不足」として
$$\vec m_{A_i, A_j}^{C_t,C_v} = (0, 0, 0, 1)$$
とします。

具体的には、以下のように考えます。

|物質|$A_i$|$A_j$|X|?|
|--|--|--|--|--|
|AgCu|Ag|Cu|$\emptyset$|無|
|AgFeCu|Ag|Cu|Fe|無|
|AgMgCu|Ag|Cu|Mg|無|
|AgMgRhCu|Ag|Cu|FeRh|不明|
|AgPdRhCu|Ag|RhCu|Pd|有|

ここで不明は観測データが無い、という意味です。
そして、$A_i$と$A_j$に対して観測データを証拠としてDempsterの規則で証拠を統合し、
元素組み合わせ、$C_t$と$C_v$の間の信念割りて関数
$\vec m^{C_t,C_v}$
を得ます。
ここで、
$\vec m = ( m(\emptyset)), m(\{類似\}), m(\{非類似\}), m(\{類似,非類似\}))$の順です。

元素組み合わせ間の類似度が信念割当関数の形で計算できました。

#### 2. 存在する・しない固溶体から元素組み合わせを置換した固溶体の信念割当関数

既存物質$A_k$から元素組み合わせ$C_t$を
$C_v$とすると物質$A_l$になるとします。

置換前物質$A_k$が存在する場合の信念割当関数を$(0, s, 0, 1-s)$ として、
この置換操作による信念割当関数を
$$\vec m_{A_k}^{A_l}(C_t,C_v) = ( 0, s \times m^{C_t, C_v}(\{類似\}), 0, 1-s \times m^{C_t, C_v}(\{類似\}))$$
とします。ここで、
$$\vec m = ( m(\emptyset)), m(\{存在有\}), m(\{存在無\}), m(\{存在有,存在無\}))$$の順序です。

また、
置換前物質$A_k$が存在しない場合の信念割当関数を$(0, 0,s, 1-s)$として、
この置換操作による$A_l$の信念割当関数を
$$\vec m_{A_k}^{A_l}(C_t,C_v) = ( 0, 0, s \times m^{C_t, C_v}(\{類似\}), 1-s \times m^{C_t, C_v}(\{類似\}))$$
とします。

$s$は$0<s<1$はデータからの学習で最適化するパラメタです。
観測データの範囲内で、$A_l$に至る$(A_k, C_t,C_v)$の経路は複数あります。これらを証拠理論で統合します。

なお、2番目の過程で、$m^{C_t, C_v}(\{類似\})$しか用いていないのは「新帰納法」として様々な推論方法を試した結果です。
