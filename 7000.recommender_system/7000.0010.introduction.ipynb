## 解説

協調フィルタリング

例えば,
A$_x$B$_{1-x}$という物質を考える。
A=Liの組成xを0.1刻みで既存の実験結果を文献からまとめると以下の表になるとします。
灰色部分がその物質が存在する領域だとします。
![problem](image_keep/collaborative_filtering.png)

物理の知識があると、LiとNaは似ているが分かります。
Liの作成可能・作成不能な物質はNaでも同様に作成可能・作成不能では無いかと期待したくなります。
協調フィルタリングを用いてこの操作を行うことができます。

存在だけでなく、classificationの目的変数のように何かの値以上を持つ物質としても良く、データがあれば適用範囲は大きいはずです。

具体的には

1. 説明変数空間全体を何かの操作で少ない特徴ベクトルに縮約する。（次元圧縮する。）
2. その少ない特徴ベクトルから１で使った操作を逆に用いて再び説明変数空間全体に展開し直す。(低ランク近似）

![](image_keep/dim_red.png)

という作業を行います。
少ない特徴ベクトルが系の特徴（法則）を十分に表しているすれば、何かしらの「規則」が現れることが期待されます。
