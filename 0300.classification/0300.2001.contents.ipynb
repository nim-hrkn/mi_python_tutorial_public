## 例題２


Q. 解明しようしている材料問題（仮想問題）が何か？

A. 単一元素からなる基底状態の結晶を機械学習で予測する。

Q. データがどうようなものか？

A. 説明変数：それぞれの構成元素の特徴量

A. 目的変数：安定構造

Q. どのような手法で何を抽出しようとするか？

A. logistic regressionによりclassificationを行う。
