## 例題１


Q. 解明しようしている材料問題（仮想問題）が何か？

A. bcc,fcc,hcp Feの構造を人間が目で見て分類するのではなくデータ科学によるクラスタリングを行う。

Q. データがどうようなものか？

A. bcc,fcc,hcp Feの構造データをBehlerの二体symmetry functionによる変換した量を記述子とする。

Q. どのような手法で何を抽出しようとするか？

A. KMeans,GMMによるクラスタリングと階層クラスタリングを行う。
