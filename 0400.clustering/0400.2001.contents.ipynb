## 例題２


Q.　解明しようしている材料問題（仮想問題）が何か？

A. 炭素８原子からなる構造を人間が目で見て分類するのではなくデータ科学による自動的な分類を行う。

Q. データがどうようなものか？

A. GRRM法により構造探索を行った炭素８原子による構造をBehlerの二体symmetry functionによる変換した量を説明変数とする。

Q. どのような手法で何を抽出しようとするか？

A. KMeans,GMM,階層クラスタリングによる構造のclusteringを行う。
