
## 例題


Q. 解明しようしている材料問題（仮想問題）が何か？

A.　物性量の予測を行う。
ただし、物性量の予測値だけでなく、予測値の標準偏差を求める。

Q. データがどうようなものか？


A.1. 簡単な線形モデルサンプル。

A.2. 希土類Co合金の磁気相転移温度データ。

A.2. 非線形モデルMorseポテンシャルに従い、測定誤差を含むデータ


Q. どのような手法で何を抽出しようとするか？

A. MCMCを用いたベイズ回帰により平均値と標準偏差を求める。
