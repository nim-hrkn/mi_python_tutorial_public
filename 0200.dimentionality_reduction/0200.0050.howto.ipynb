## scikit-learnクラスの説明

3次元に次元圧縮する場合の例を示します。

- PCA
```python
    from sklearn.decomposition import PCA
    pca = PCA(3)
    pca.fit(X)
    X_pca　=　pca.transform(X)
```

- MDS
```python
    from sklearn.manifold import MDS
    red = MDS(3)
    X_red = red.fit_transform(X)
```

- t-SNE
```python
    from sklearn.manifold import TSNE
    red = TSNE(3)
    X_red = red.fit_transform(X)
```
