### 説明変数作成

元データでは数が大きすぎるので、名前が付けたデータのみに制限して、csvをセーブします。


```python
import pandas as pd

g_df0 = pd.read_csv(
    "../data_calculated/Carbon8_descriptor.csv", index_col=[0, 1])


def select_structure(df):
    """selecte data from DataFrame

    Args:
        df (pd.DataFrame): data

    Returns:
        pd.DataFrame: selected data
    """
    keylist = []
    for key in df.index:
        s = key[0].split("-")
        dim = s[0]
        i = int(s[1])
        if (dim == "3D" or dim == "2D" or dim == "1D") and i <= 5:
            keylist.append(key)
    return df.loc[keylist]


g_df_desc = select_structure(g_df0)
g_df_desc
```

```python
g_df1 = pd.read_csv(
    "../data_calculated/carbon8_nn_atom_longform.csv", index_col=[0])
g_df_polytype = g_df1.set_index(["key", "atom"])
g_df_polytype
```

```python
g_df = pd.concat([g_df_desc, g_df_polytype], axis=1).dropna()
g_filename = "../data_calculated/Carbon8_descriptor_selected.csv"
g_df.to_csv(g_filename)
```
