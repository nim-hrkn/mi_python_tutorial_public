
## 解説


### ノイズを含む観測値

観測値 
$$
t_i = f(x_i) + \epsilon
$$
は$f(x_i)$で決まる予測値と$\epsilon$で決まるノイズを含とします。
このノイズ$\epsilon$が正規分布に従うとします。つまり、
$
t_i \sim N(f(x_i), \sigma^2)
$ です。
ここに$\sigma^2$は正規分布の標準偏差です。
$f(x_i)$の分布は
$$
P(f(x_i)) = \frac{1}{\sqrt{(2\pi)\sigma^{2}}} \exp\left( -\frac{1}{2\sigma^2} (t_i -f(x_i) )^2  \right) 
$$
に従います。

予測モデルとして
線形回帰モデル
$$
f(x) = \sum_p c_p \phi_p(x) 
$$
を考えます。

### ベイズ推定による係数評価

確率分布が$P(f(x_i))$
で与えられる場合に他に何ができるでしょうか。
係数に関してもやはり線形な式で$c_p$は$x_i$に依らないので
expの中は係数$c_p$の二次形式にもなっています。係数$c_p$に関して最安定解($c^{min}$)からゆらぎ（共分散行列）を解析的に計算することが可能です。
