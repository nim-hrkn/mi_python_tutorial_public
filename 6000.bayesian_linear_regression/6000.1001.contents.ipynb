## 例題

Q. 解明しようとしている材料問題（仮想問題）が何か？

A. 物性量の予測を行う。
ただし、物性量の予測値だけでなく、予測値の標準偏差を求める。

Q. データはどのようなものか？

A.1. 簡単な例
    * 説明変数: xと sin関数
    * 目的変数: sin関数

A.2. ReCo結晶

    * 説明変数: ReCo結晶説明変数
    * 目的変数: ReCo結晶キュリー温度


A.3. ZWとRS結晶

    * 説明変数: ZWとRS結晶説明変数
    * 目的変数: ZWとRS結晶のエネルギー差


Q. どのような手法で何を抽出しようとするか？

A. 線形ベイズ回帰により平均値と標準偏差を求める。
