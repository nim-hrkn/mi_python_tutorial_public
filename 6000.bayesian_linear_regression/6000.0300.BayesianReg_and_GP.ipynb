# ベイズ線形回帰とガウシアンプロセス回帰の共通点と相違点

## 1. 共通点：ベイズ更新

ベイズ線形回帰（Bayesian Linear Regression）とガウシアンプロセス回帰（Gaussian Process Regression, GP）は、どちらも**ベイズ推論**に基づいています。

どちらも

* 事前分布（Prior）
  $$
  P(\theta)
  $$
* 尤度（Likelihood）
  $$
  P(D|\theta)
  $$

から

* 事後分布（Posterior）
  $$
  P(\theta|D)
  $$

を計算します。

つまり、

> **データを見る前の信念（Prior）を、観測データによって更新してPosteriorを得る**

という考え方は共通です。

## 2. ベイズ線形回帰

### モデル

ベイズ線形回帰では、$\phi(x)$にintercept項を含めて

$$
y=\mathbf{w}^{T}\boldsymbol{\phi}(x)+\varepsilon
$$

を仮定します。

ここで

* $\mathbf{w}$：回帰係数ベクトル
* $\boldsymbol{\phi}(x)$：基底関数ベクトル
* $\varepsilon\sim N(0,\sigma_n^2)$：ガウスノイズ

です。

未知なのは

$$
\mathbf{w}
$$

です。

---

### Prior

重みに対して

$$
\mathbf{w}\sim N(\boldsymbol{\mu}_w,\Sigma_w)
$$

という事前分布を置きます。

ここで

* $\boldsymbol{\mu}_w$：重みの平均ベクトル
* $\Sigma_w$：重みの共分散行列

です。

例えば、Prior
$$
\boxed{
\mathbf{w}\sim N(0,\alpha^{-1}I)
}
$$
は

* 全ての重みは平均0
* 互いに独立
* 同じ分散

という仮定です。


---

### $\mu_w$ と $\Sigma_w$ の意味

#### 期待値

ある確率分布 $p(\mathbf w)$ に従う確率変数 $\mathbf w$ と、関数 $g(\mathbf w)$ に対して、期待値を

$$
E_{p(\mathbf w)}[g(\mathbf w)]
=

\int g(\mathbf w)p(\mathbf w)d\mathbf w
$$

と定義します。

多くの教科書では、分布 $p(\mathbf w)$ が明らかな場合には省略して

$$
E[g(\mathbf w)]
$$

と書きます。

---

#### 平均ベクトル

関数

$$
g(\mathbf w)=\mathbf w
$$

とすると、

$$
\boxed{
E[\mathbf w]=\boldsymbol{\mu}_w
}
$$

となります。

例えば

$$
\boldsymbol{\mu}_w=
\begin{pmatrix}
0\
0\
0
\end{pmatrix}
$$

なら、

「どの重みも平均的には0付近だろう」

という事前知識を表します。

---

#### 共分散行列

共分散行列は

$$
\boxed{
\Sigma_w
=

E\left[
(\mathbf w-\boldsymbol{\mu}_w)
(\mathbf w-\boldsymbol{\mu}_w)^T
\right]
}
$$

で定義されます。

これは、重みベクトルが平均からどの程度ばらつくかを表します。

例えば

$$
\Sigma_w=I
$$

なら、

各重みは互いに独立で、分散が1です。

一方、

$$
\Sigma_w=
\begin{pmatrix}
1&0.8\\
0.8&1
\end{pmatrix}
$$

なら、

2つの重み $w_1,w_2$ は正の共分散を持ち、正の相関があります。


---

### Posterior

データ

$$
D=(X,y)
$$

を観測すると

$$
P(\mathbf w|D)
$$

が計算できます。

つまり**重みの確率分布**が更新されます。

---

### 予測

新しい入力

$$
x_*
$$

では

$$
P(y_*|x_*,D)
$$

を求めます。

これは**重みの不確実性を積分した予測分布**です。

---


### ガウシアンプロセス(GP)の定義


任意の有限個の入力

$$
x_1,\dots,x_N
$$

に対して、関数値ベクトル

$$
\mathbf f=
\begin{pmatrix}
f(x_1)\\
\vdots\\
f(x_N)
\end{pmatrix}
$$

が、平均ベクトル (\mathbf m)、共分散行列 (K) をもつ多変量ガウス分布

$$
\boxed{
\mathbf f\sim N(\mathbf m,K)
}
$$

に従うとき、

$$
\boxed{
f(x)\sim GP(m,k)
}
$$

と書きます。

ここで、

$$
\mathbf m=
\begin{pmatrix}
m(x_1)\\
\vdots\\
m(x_N)
\end{pmatrix},
$$

$$
K_{ij}=k(x_i,x_j)
$$

です。



### 多変量正規分布 (N) と ガウシアンプロセス(GP) の違い

多変量正規分布とガウシアンプロセス(GP)はどちらもガウス分布に基づいていますが、**対象とする確率変数が異なります。**

#### 多変量正規分布

多変量正規分布では、**有限次元のベクトル**を確率変数として扱います。

$$
\boxed{
\mathbf{x}\sim N(\mathbf{m},\Sigma)
}
$$

ここで、

* $\mathbf{x}$：(N) 次元の確率変数ベクトル
* $\mathbf{m}$：平均ベクトル
* $\Sigma$：共分散行列

です。

つまり、有限個の確率変数

$$
(x_1,\ldots,x_N)
$$

の同時分布を表します。


#### GP関数

ガウシアンプロセスでは、**関数 (f(x)) 全体**を確率変数として扱います。

$$
\boxed{
f(x)\sim GP(m(x),k(x,x'))
}
$$

ここで、

* $m(x)$：平均関数
* $k(x,x')$：共分散関数（カーネル）

です。


GPは、

> **任意の有限個の入力点**

$$
x_1,\ldots,x_N
$$

を選ぶと、

$$
\mathbf f=
\begin{pmatrix}
f(x_1)\\
\vdots\\
f(x_N)
\end{pmatrix}
$$

が

$$
\boxed{
\mathbf f\sim N(\mathbf m,K)
}
$$

となるような**確率過程**として定義されます。

ここで、

$$
\mathbf m=
\begin{pmatrix}
m(x_1)\
\vdots\
m(x_N)
\end{pmatrix},
\qquad
K_{ij}=k(x_i,x_j)
$$

です。


#### 両者の関係

GPは**関数全体の分布**ですが、その関数を有限個の入力点で評価すると、必ず**多変量正規分布**になります。

すなわち、

$$
\boxed{
f(x)\sim GP(m,k)
\quad\Longrightarrow\quad
(f(x_1),\ldots,f(x_N))
\sim N(\mathbf m,K)
}
$$

となります。

## 3. Gaussian Process回帰

### 定義
- Gaussian Process（GP）：関数に対する **事前分布（Prior）** を表す確率過程。

- Gaussian Process Regression（GPR）：GPを事前分布として用い、観測データから **事後分布（Posterior）** を求めて回帰を行う手法。

---
### GP回帰

GP回帰では

$$
y=f(x)+\varepsilon
$$

と書きます。未知なのは

$$
f(x)
$$

という**関数そのもの**です。

---

### Prior

GPでは

$$
\boxed{
f(x)\sim GP(m(x),k(x,x'))
}
$$

を仮定します。

ここで

* $m(x)$：平均関数
* $k(x,x')$：共分散関数（カーネル）

です。

---

### Posterior

データを観測すると

$$
P(f|D)
$$

になります。

つまり

**関数そのもの**

のPosteriorです。

---

### 予測

新しい点

$$
x_*
$$

では

$$
f(x_*)
$$

の

* 平均
* 分散

が直接求まります。

>この過程で具体的に$m$と$k$が求まります。

## 4. ベイズ線形回帰との対応


$$
f(x)
$$

は

$$
\boxed{
f(x)\sim GP\left(
\phi(x)^T\mu_w,;
\phi(x)^T\Sigma_w\phi(x')
\right)
}
$$

となります。

これが

**ベイズ線形回帰をGPで書いた式**


したがって,ベイズ線形回帰はGPの特殊な場合とみなせます。

## 5. RBF(Gaussian) Kernelとの関係



### Mercer展開

RBF Kernel

$$
k(x,x')
=

\exp
\left(
-\frac{(x-x')^2}{2l^2}
\right)
$$

は

$$
(x-x')^2
=

x^2
-2xx'
+x'^2
$$

より

$$
k(x,x')
=

e^{-x^2/2l^2}
e^{-x'^2/2l^2}
e^{xx'/l^2}
$$

中央をTaylor展開すると

$$
e^{xx'/l^2}
=

\sum_{n=0}^{\infty}
\frac{1}{n!}
\left(
\frac{xx'}{l^2}
\right)^n
$$

したがって

$$
k(x,x')
=

\sum_{n=0}^{\infty}
\phi_n(x)\phi_n(x')
$$

となります。

ここで

$$
\boxed{
\phi_n(x)
=

e^{-x^2/2l^2}
\frac{x^n}{l^n\sqrt{n!}}
}
$$

です。

つまり

RBF Kernelは

**無限次元の特徴空間の内積**

です。

## 6. パラメトリックとノンパラメトリック

### ベイズ線形回帰（Parametric）

未知なのは

$$
\mathbf w
$$

だけです。

基底関数が10個なら未知パラメータも10個です。
データ数が増えても未知数は増えません。

| データ数  | 重み数 |
| ----- | --- |
| 20    | 10  |
| 100   | 10  |
| 10000 | 10  |

---

### Gaussian Process（Nonparametric）

未知なのは

$$
f(x)
$$

です。学習点が$N$個ならPosteriorは

$$
(f(x_1),\ldots,f(x_N))
$$

という$N$次元ガウス分布になります。

したがって

| データ数  | 推定対象の自由度 |
| ----- | -------- |
| 20    | 20       |
| 100   | 100      |
| 10000 | 10000    |

となります。

ノンパラメトリックとは
**パラメータが無い**という意味ではなく
**モデルの自由度がデータ数とともに増える**という意味です。

## 7. GPのパラメータ

例えばRBF Kernel

$$
k(x,x')
=

\sigma_f^2
\exp
\left(
-\frac{|x-x'|^2}{2l^2}
\right)
$$

には

* 長さ尺度 $l$
* 信号分散 $\sigma_f^2$
* ノイズ分散 $\sigma_n^2$

などの**ハイパーパラメータ**があります。

これらは通常、周辺尤度（marginal likelihood）の最大化などにより推定します。したがって

> **ノンパラメトリック＝パラメータが存在しない**

ではありません。

## 8. 平均関数 $m(x)$ の $x\to\pm\infty$ の挙動

多くの教科書では、ガウス過程は

$$
m(x)=0
$$

を仮定します。つまり

$$
f(x) \sim GP(0, k)
$$

です。

その場合、RBFカーネルを用いたGaussian Processは、観測点から十分離れた入力ではカーネルの値が0、つまり

$$
k(x,x_i)\rightarrow 0
$$

となるため、事後平均は事前平均$m(x)$の極限値、

$$
\boxed{
x\rightarrow\pm\infty
\quad\Longrightarrow\quad
f(x)\rightarrow m(x)=0
}
$$

となります。


> **Gaussian Processは $x\to\infty$ で予測平均が0になるのではなく。事前平均関数 $m(x)\equiv 0$ を仮定している場合に、$x\to\infty$ で予測平均が0となります。**
