$$
\text{BCE Loss}(\mathbf{p}, \mathbf{y}) = - \frac{1}{N} \sum_{i=1}^N \left[ y_i \log(p_i) + (1 - y_i) \log(1 - p_i) \right] 
$$

$y_i$は0または1の値を取る．
$y_i$が1の時は第一項，0の時は第二項を用いて評価する．

最小化する場合，マイナスが付いているので
第一項は，$p_i$を1に近づける．
第二項は，$p_i$を0に近づける．



```python
import torch
import torch.nn as nn

# Binary Cross-Entropy Loss functionのインスタンスを作成
criterion = nn.BCELoss()

# ユーザーが指定する任意のベクトル v (例として [0, 1, 0, 1, 1, 0])
v = torch.tensor([0, 0, 1, 0, 0, 0], dtype=torch.float32)

# ベクトル v と同じ長さで全ての要素が 1 のベクトルを作成
one = torch.ones_like(v)

# ベクトル v と同じ長さで全ての要素が 0 のベクトルを作成
zero = torch.zeros_like(v)

# Binary Cross-Entropy Lossの計算
loss_v_one = criterion(v, one)
loss_v_zero = criterion(v, zero)

# 計算結果を表示
print(f"ユーザー定義のベクトル: {v}")
print(f"全ての要素が1のベクトル: {one}")
print(f"全ての要素が0のベクトル: {zero}")
print(f"criterion(v, one)の損失: {loss_v_one.item()}")
print(f"criterion(v, zero)の損失: {loss_v_zero.item()}")
```
