# SMILES

分子を直線と分岐と、直線にするための切断位置とで表す手法。
細かい部分には方言がある。

以下には簡単な例を示す。

```python
from rdkit import Chem
from rdkit.Chem import Draw
from rdkit.Chem.Draw import IPythonConsole  # Jupyter 表示用
```

```python
# 直線状分子, エタノール

smiles = "CCO"
mol = Chem.MolFromSmiles(smiles)

mol 
```

```python
# 更に長く
# 直線状分子,1-プロパノール（プロピルアルコール）

smiles = "CCCO"
mol = Chem.MolFromSmiles(smiles)

mol 
```

```python
# 途中で分岐
# 2-(アミノメチル)-1-プロパノール

smiles = "CC(CN)CO"

mol = Chem.MolFromSmiles(smiles)

mol
# に２つめのC元素にCOHを分岐させる
```

```python
# 芳香族の書き方、メチルベンゼン
# 芳香族は小文字にする。
# c1-c1同士を接続する。

smiles = "Cc1ccccc1"
mol = Chem.MolFromSmiles(smiles)

mol 
```

```python
# 芳香族,o-キシレン
# ortho位置

smiles = "Cc1c(C)cccc1"
mol = Chem.MolFromSmiles(smiles)

mol 
```

```python
# 芳香族, m-キシレン
# meta 位置

smiles = "Cc1cc(C)ccc1"
mol = Chem.MolFromSmiles(smiles)

mol 
```

```python
# 芳香族, p-キシレン
# para位置

smiles = "Cc1ccc(C)cc1"
mol = Chem.MolFromSmiles(smiles)

mol 
```

```python
# 1,2-ジクロロエチレンのcis,transの区別

# cis-1,2-ジクロロエチレン

smiles = "Cl/C=C\Cl"

mol = Chem.MolFromSmiles(smiles)

mol
```

```python
# trans-1,2-ジクロロエチレン

smiles = "Cl/C=C/Cl"
mol = Chem.MolFromSmiles(smiles)

mol 
```
